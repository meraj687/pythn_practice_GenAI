````python

import random

com = random.randint(1, 100)
attempts = 5

print("🎮 Welcome to Number Guessing Game!")
print("I have selected a number between 1 and 100.")
print("You have 5 attempts to guess it.\n")

while attempts > 0:
    human = int(input("Enter your guess: "))

    if human == com:
        print("🎉 Congratulations! You won!")
        break

    elif human > com:
        print("Guess a lower number!")

    else:
        print("Guess a higher number!")

    attempts -= 1
    print("Attempts remaining:", attempts, "\n")

if attempts == 0 and human != com:
    print("😢 You lost!")
    print("The correct number was:", com)
````

````python

import random

com = random.randint(1, 100)
attempts = 5

print("Welcome to Number Guessing Game!")
print("I have selected a number between 1 and 100.")
print("You have 5 attempts to guess it.\n")

while attempts > 0:
    guess = input("Enter your guess: ")

    if not guess.isdigit():
        print("Please enter a valid number!\n")
        continue

    human = int(guess)

    if human == com:
        print("Congratulations! You won!")
        break

    elif human > com:
        print("Guess a lower number!")

    else:
        print("Guess a higher number!")

    attempts -= 1
    print("Attempts remaining:", attempts, "\n")

if attempts == 0 and human != com:
    print("You lost!")
    print("The correct number was:", com)
````

````python
def palindrome(st):
  rev = ""
  for i in range(len(st)-1,-1,-1):
    rev = rev + st[i]

  if rev == st:
    print(f" {st} Palindrom")
  else:
    print(f" {st} Not palidrom")

palindrome("NAMAN")
palindrome("ARYAAN")
````

````python
def chat_covnersion(text):
  new_text = []
  for w in text.split():
    if w.upper() in chat_words:
      new_text.append(chat_words[w.upper()])
    else:
      new_text.append(w)
  return " ".join(new_text)
````

````python
def remove_stopwords(text):
    new_text = []

    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)
````

````python
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
````

````python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset("TimSchopf/medical_abstracts")

print(dataset)
````

````python
train_df = dataset["train"].to_pandas()
test_df = dataset["test"].to_pandas()

print("Training data:", train_df.shape)
print("Testing data:", test_df.shape)

print(train_df.head())
print(train_df.columns.tolist())
````

````python
print(train_df.isnull().sum())
print(train_df["condition_label"].value_counts())

# Display one medical abstract
print(train_df["medical_abstract"].iloc[0])
````

````python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from bs4 import BeautifulSoup
import re

stop_words = set(stopwords.words("english"))

# Preserve medically important negations
stop_words -= {"no", "nor", "not"}

lemmatizer = WordNetLemmatizer()
````

````python
def preprocess_text(text):

    if pd.isna(text):
        return ""

    # Remove HTML tags
    text = BeautifulSoup(str(text), "html.parser").get_text(" ")

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Remove punctuation, digits and special characters
    text = re.sub(r"[^a-z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenization
    tokens = word_tokenize(text)

    # Stopword removal
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # Lemmatization
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    return " ".join(tokens)
````

````python
original_text = train_df["medical_abstract"].iloc[0]

cleaned_text = preprocess_text(original_text)

print("ORIGINAL ABSTRACT:\n")
print(original_text[:1000])

print("\nPREPROCESSED ABSTRACT:\n")
print(cleaned_text[:1000])
````

````python
sample_df = train_df.head(500).copy()

sample_df["cleaned_abstract"] = (
    sample_df["medical_abstract"].apply(preprocess_text)
)

sample_df[
    ["medical_abstract", "cleaned_abstract", "condition_label"]
].head()
````

````python
train_df["cleaned_abstract"] = (
    train_df["medical_abstract"].apply(preprocess_text)
)

test_df["cleaned_abstract"] = (
    test_df["medical_abstract"].apply(preprocess_text)
)

print("Training preprocessing completed!")
print("Testing preprocessing completed!")
````

````python
i = 0

print("BEFORE:\n")
print(train_df["medical_abstract"].iloc[i])

print("\nAFTER:\n")
print(train_df["cleaned_abstract"].iloc[i])

print("\nOriginal word count:",
      len(word_tokenize(train_df["medical_abstract"].iloc[i])))

print("Cleaned word count:",
      len(word_tokenize(train_df["cleaned_abstract"].iloc[i])))
````

````python
train_df["split"] = "train"
test_df["split"] = "test"

final_df = pd.concat(
    [train_df, test_df],
    ignore_index=True
)

final_df.to_csv(
    "medical_abstracts_preprocessed.csv",
    index=False
)

print("Total records:", len(final_df))
print("File saved successfully!")
````

````python
import nltk
import re

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

stop_words = set(stopwords.words("english"))

# Keep important negations in medical text
stop_words -= {"no", "nor", "not"}

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
````

````python
def clean_text(text):
    text = str(text).lower()

    # Remove punctuation, digits and special characters
    text = re.sub(r"[^a-z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
````

````python
sample = "I have FEVER, headache, and body pain!"

print("Before:", sample)
print("After:", clean_text(sample))
````

````python
sample = "I have fever and headache"

tokens = word_tokenize(sample.lower())

print(tokens)
````

````python
def remove_stopwords(text):
    tokens = word_tokenize(text)

    filtered_tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    return filtered_tokens

print(remove_stopwords("I have a fever and headache"))
````

````python
words = ["studies", "playing", "cars"]

print("Original:", words)

print(
    "Stemmed:",
    [stemmer.stem(word) for word in words]
)

print(
    "Lemmatized:",
    [lemmatizer.lemmatize(word) for word in words]
)
````

````python
def preprocess_text(text):
    # Handle missing values
    if pd.isna(text):
        return ""

    # Clean the text
    text = clean_text(text)

    # Tokenize
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # Lemmatize
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    # Join words into a single string
    return " ".join(tokens)
````

````python
# Find the symptom description column
text_column = next(
    (col for col in ["text", "symptom", "symptoms", "description"]
     if col in df.columns),
    None
)

if text_column is None:
    raise ValueError(
        f"Cannot identify text column. Available columns: {df.columns.tolist()}"
    )

df["cleaned_text"] = df[text_column].apply(preprocess_text)

display(df[[text_column, "cleaned_text"]].head())
````

````python
i = 0

print("ORIGINAL TEXT:\n")
print(df[text_column].iloc[i])

print("\nCLEANED TEXT:\n")
print(df["cleaned_text"].iloc[i])

print("\nOriginal word count:",
      len(word_tokenize(clean_text(df[text_column].iloc[i]))))

print("Cleaned word count:",
      len(word_tokenize(df["cleaned_text"].iloc[i])))
````

````python
output_file = "symptom2disease_preprocessed.csv"

df.to_csv(output_file, index=False)

print("File saved:", output_file)
print("Total records:", len(df))
````

````python
import os
from nltk.tokenize import sent_tokenize
from gensim.utils import simple_preprocess

data_path = "/kaggle/input/game-of-thrones-book-dataset"

story = []

for filename in os.listdir(data_path):
    if filename.endswith(".txt"):
        file_path = os.path.join(data_path, filename)

        with open(file_path, "r", encoding="cp1252") as f:
            corpus = f.read()

        raw_sent = sent_tokenize(corpus)

        for sent in raw_sent:
            story.append(simple_preprocess(sent))

print("Total sentences:", len(story))
print("First 5 sentences:", story[:5])
````

````python

import re

def clean_text(text):
    text = text.lower()  # Convert to lowercase
    text = re.sub(r"<br\s*/?>", " ", text)  # Remove HTML line breaks
    text = re.sub(r"http\S+|www\S+", " ", text)  # Remove URLs
    text = re.sub(r"[^a-z\s']", " ", text)  # Keep letters and apostrophes
    text = re.sub(r"\s+", " ", text).strip()  # Remove extra spaces
    return text

# Test on one review first
original_review = df["review"].iloc[0]
cleaned_review = clean_text(original_review)

print("ORIGINAL REVIEW:\n")
print(original_review[:500])

print("\nCLEANED REVIEW:\n")
print(cleaned_review[:500])
````

````python

df["clean_review"] = df["review"].apply(clean_text)

print(df[["review", "clean_review", "sentiment"]].head(3))
````

````python

from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))

# Keep negation words because they matter for sentiment
stop_words = stop_words - {"not", "no", "nor"}

sample_tokens = word_tokenize(
    "this movie was not good and the acting was excellent"
)

filtered_tokens = [
    word for word in sample_tokens
    if word not in stop_words
]

print("Original tokens:")
print(sample_tokens)

print("\nAfter stopword removal:")
print(filtered_tokens)
````

````python

my_review = "The movie was not interesting, but the acting was excellent!"

# Clean the review
cleaned = clean_text(my_review)

# Tokenize it
my_tokens = word_tokenize(cleaned)

# Remove stopwords while preserving negations
my_filtered_tokens = [
    word for word in my_tokens
    if word not in stop_words
]

print("Cleaned review:")
print(cleaned)

print("\nTokens:")
print(my_tokens)

print("\nFiltered tokens:")
print(my_filtered_tokens)
````

````python

from sklearn.feature_extraction.text import CountVectorizer

sample_reviews = [
    "this movie is good",
    "this movie is bad"
]

bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(sample_reviews)

print("Vocabulary:")
print(bow_vectorizer.get_feature_names_out())

print("\nBoW matrix:")
print(bow_matrix.toarray())
````

````python

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(sample_reviews)

print("Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF matrix:")
print(tfidf_matrix.toarray().round(3))
````

````python

reviews = [
    "This movie was fantastic",
    "This movie was terrible",
    "The movie was fantastic and fantastic"
]

bow = CountVectorizer()
bow_result = bow.fit_transform(reviews)

tfidf = TfidfVectorizer()
tfidf_result = tfidf.fit_transform(reviews)

print("BoW features:")
print(bow.get_feature_names_out())
print(bow_result.toarray())

print("\nTF-IDF features:")
print(tfidf.get_feature_names_out())
print(tfidf_result.toarray().round(3))
````

````python

from sklearn.model_selection import train_test_split

# Convert labels to numbers
df["label"] = df["sentiment"].map({
    "positive": 1,
    "negative": 0
})

print(df[["sentiment", "label"]].head())
print("\nLabel counts:")
print(df["label"].value_counts())
````

````python

X_train, X_test, y_train, y_test = train_test_split(
    df["clean_review"],
    df["label"],
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)

print("Training reviews:", len(X_train))
print("Testing reviews:", len(X_test))
````

````python

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(
    max_features=20000,
    ngram_range=(1, 2),
    min_df=2
)

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

print("Training matrix shape:", X_train_tfidf.shape)
print("Testing matrix shape:", X_test_tfidf.shape)
````

````python

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train_tfidf, y_train)

print("Model training completed!")
````

````python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

y_pred = model.predict(X_test_tfidf)

print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Negative", "Positive"]
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
````

````python

custom_reviews = [
    "This movie was absolutely fantastic and entertaining",
    "This was a terrible movie and I hated it",
    "The movie was not good"
]

custom_tfidf = tfidf_vectorizer.transform(
    [clean_text(review) for review in custom_reviews]
)

predictions = model.predict(custom_tfidf)
probabilities = model.predict_proba(custom_tfidf)

for review, prediction, probability in zip(
    custom_reviews, predictions, probabilities
):
    sentiment = "Positive" if prediction == 1 else "Negative"
    confidence = probability[prediction]

    print("Review:", review)
    print("Predicted sentiment:", sentiment)
    print("Model probability for prediction:", round(confidence, 3))
    print("-" * 50)
````

````python
from transformers import pipeline

texts = [
    "I really liked the autoencoder.",
    "I am worried that my AI project will fail."
]

# Model 1: Zero-shot classification
zero_shot = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

labels = ["positive", "negative", "neutral"]

for text in texts:
    result = zero_shot(text, candidate_labels=labels)
    print("\nText:", text)
    print("BART labels:", result["labels"][:3])
    print("BART scores:", result["scores"][:3])

# Model 2: Emotion classification
emotion_model = pipeline(
    "text-classification",
    model="SamLowe/roberta-base-go_emotions",
    top_k=None
)

results = emotion_model(texts)

for text, emotions in zip(texts, results):
    top3 = sorted(
        emotions,
        key=lambda x: x["score"],
        reverse=True
    )[:3]

    print("\nText:", text)
    print("RoBERTa emotions:", top3)
````

````python
def convert_examples_to_features(example_batch):
  input_encodings = tokenizer(example_batch["dialogue"] , max_length = 1024 , truncation = True)

  with tokenizer.as_target_tokenizer():
    target_encodings = tokenizer(example_batch["summary"] , max_length =1024 , truncation = True)
  return {
      "input_ids" : input_encodings["input_ids"] ,
      "attention_mask" : input_encodings["attention_mask"] ,
      "labels" : target_encodings["input"]}
````

````python

def convert_examples_to_features(example_batch):

    # Tokenize conversations
    input_encodings = tokenizer(
        example_batch["dialogue"],
        max_length=512,
        truncation=True
    )

    # Tokenize summaries without as_target_tokenizer()
    target_encodings = tokenizer(
        text_target=example_batch["summary"],
        max_length=128,
        truncation=True
    )

    return {
        "input_ids": input_encodings["input_ids"],
        "attention_mask": input_encodings["attention_mask"],
        "labels": target_encodings["input_ids"]
    }
````

````python
def generate_batch_sized_chunks(list_of_elements, batch_size):
    """split the dataset into smaller batches that we can process simultaneously
    Yield successive batch-sized chunks from list_of_elements."""
    for i in range(0, len(list_of_elements),batch_size):
      yield list_of_elements[i : i + batch]


def calculate_metrics_on_test(dataset , metrics , model , tokenizer , batch_size = 16 , device = device , column_text = "article" , column_summary = "highlights"):
  article_batches = list(generate_batch_sized_chunks(dataset[column_text] , batch_size))
  target_batches = list(generate_batch_sized_chunks(dataset[column_summary] , batch_size))

  for article_batch , target_batch in tqdm(
      zip(article_batches , target_batches) , total = len(article_batches)
  ):
    inputs = tokenizer(article_batch , max_length = 1024 , truncation = True , padding = "max_length" , return_tensors = "pt")
    summaries = model.generate(input_ids = inputs["input_ids"].to(device) , attention_mask = inputs["attention_mask"].to(device) , length_penalty = 0.8 , num_beams = 8 , max_length = 128)
    """parameter forlength"""

    decoded_summaries = [d.replace("<pad>" , " ") for d in tokenizer.batch_decode(summaries , skip_special_tokens = True , clean_up_tokenization_spaces = True)]

    decoded_summaries = [d.replace("" , " ") for d in decoded_summaries]

    metric.add_batch(prediction = decoded_summaries , reference = target_batch)

  score = metric.compute()
  return score
````

````python

# Find potential summarization examples
summary_examples = []

for example in dataset:
    question = str(example.get("question", ""))
    system_prompt = str(example.get("system_prompt", ""))

    text_to_check = (question + " " + system_prompt).lower()

    if any(keyword in text_to_check for keyword in [
        "summarize", "summarise", "write a summary",
        "provide a summary", "brief summary"
    ]):
        summary_examples.append(example)

    if len(summary_examples) >= 10:
        break

print("Potential summarization examples found:", len(summary_examples))

for i, example in enumerate(summary_examples):
    print(f"\n--- Example {i + 1} ---")
    print("QUESTION:", str(example["question"])[:500])
    print("RESPONSE:", str(example["response"])[:500])
````

````python

from rouge_score import rouge_scorer
import torch
import pandas as pd

scorer = rouge_scorer.RougeScorer(
    ["rouge1", "rouge2", "rougeL"],
    use_stemmer=True
)

results = []

# Evaluate up to 10 candidate examples
for i in range(min(10, len(df_summary))):

    source_text = str(df_summary.loc[i, "question"])
    reference = str(df_summary.loc[i, "response"])

    inputs = tokenizer(
        source_text,
        return_tensors="pt",
        max_length=1024,
        truncation=True
    ).to(device)

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=100,
            num_beams=4,
            do_sample=False
        )

    prediction = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )

    scores = scorer.score(reference, prediction)

    results.append({
        "example": i + 1,
        "rouge1_f1": scores["rouge1"].fmeasure,
        "rouge2_f1": scores["rouge2"].fmeasure,
        "rougeL_f1": scores["rougeL"].fmeasure,
        "generated_summary": prediction
    })

results_df = pd.DataFrame(results)

print("Individual results:")
display(results_df)

print("\nAverage ROUGE F1 scores:")
print(results_df[
    ["rouge1_f1", "rouge2_f1", "rougeL_f1"]
].mean().round(4))
````

````python

# Search a bounded stream for summarization examples

from datasets import load_dataset
import pandas as pd

dataset = load_dataset(
    "Open-Orca/OpenOrca",
    split="train",
    streaming=True
)

candidates = []
checked = 0
max_to_check = 5000

for example in dataset:
    checked += 1

    question = str(example.get("question", "")).strip()
    response = str(example.get("response", "")).strip()
    system_prompt = str(example.get("system_prompt", "")).strip()

    instruction = (system_prompt + " " + question).lower()

    asks_for_summary = any(phrase in instruction for phrase in [
        "summarize the following",
        "summarise the following",
        "summarize this text",
        "summarise this text",
        "provide a summary of the following",
        "write a summary of the following",
        "summarize the text below",
        "summarise the text below"
    ])

    if asks_for_summary and question and response:
        candidates.append({
            "question": question,
            "reference": response
        })

    if checked >= max_to_check:
        break

df_candidates = pd.DataFrame(candidates)

print("Examples inspected:", checked)
print("Potential summarization examples:", len(df_candidates))

display(df_candidates.head(5))
````

````python

import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Choose a pretrained summarization model
model_name = "sshleifer/distilbart-cnn-12-6"

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Use GPU if available, otherwise CPU
device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)
model.eval()

print("Model loaded successfully!")
print("Device:", device)
````

````python

# Get one dialogue and its human-written summary
dialogue = dataset["train"][0]["dialogue"]
reference_summary = dataset["train"][0]["summary"]

# Convert dialogue into tokens
inputs = tokenizer(
    dialogue,
    return_tensors="pt",
    max_length=512,
    truncation=True
).to(device)

# Generate a summary
with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=100,
        num_beams=4,
        do_sample=False
    )

# Convert tokens back into readable text
generated_summary = tokenizer.decode(
    output_ids[0],
    skip_special_tokens=True
)

print("ORIGINAL DIALOGUE:\n")
print(dialogue)

print("\nHUMAN-WRITTEN SUMMARY:\n")
print(reference_summary)

print("\nMODEL-GENERATED SUMMARY:\n")
print(generated_summary)
````

````python
from rouge_score import rouge_scorer

scorer = rouge_scorer.RougeScorer(
    ["rouge1", "rouge2", "rougeL"],
    use_stemmer=True
)

scores = scorer.score(
    reference_summary,
    generated_summary
)

print("ROUGE Evaluation Results\n")

for metric, score in scores.items():
    print(f"{metric.upper()}:")
    print(f"  Precision: {score.precision:.4f}")
    print(f"  Recall:    {score.recall:.4f}")
    print(f"  F1 Score:  {score.fmeasure:.4f}")
    print()
````

````python

from rouge_score import rouge_scorer
import torch

# Select 20 conversations from the test dataset
test_data = dataset["test"].select(range(20))

scorer = rouge_scorer.RougeScorer(
    ["rouge1", "rouge2", "rougeL"],
    use_stemmer=True
)

all_scores = {
    "rouge1": [],
    "rouge2": [],
    "rougeL": []
}

model.eval()

for i, example in enumerate(test_data):
    dialogue = example["dialogue"]
    reference = example["summary"]

    inputs = tokenizer(
        dialogue,
        return_tensors="pt",
        max_length=512,
        truncation=True
    ).to(device)

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=100,
            num_beams=4,
            do_sample=False
        )

    prediction = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )

    scores = scorer.score(reference, prediction)

    for metric in all_scores:
        all_scores[metric].append(
            scores[metric].fmeasure
        )

    print(f"Example {i + 1}")
    print("Reference:", reference)
    print("Generated:", prediction)
    print("-" * 60)

print("\nAVERAGE ROUGE F1 SCORES")

for metric, values in all_scores.items():
    print(f"{metric.upper()}: {sum(values) / len(values):.4f}")
````

````python

from torch.utils.data import DataLoader
from transformers import DataCollatorForSeq2Seq

# Small subset for our first experiment
train_data = dataset["train"].select(range(500))
validation_data = dataset["validation"].select(range(100))

def preprocess(batch):
    inputs = tokenizer(
        batch["dialogue"],
        max_length=512,
        truncation=True
    )

    targets = tokenizer(
        text_target=batch["summary"],
        max_length=100,
        truncation=True
    )

    inputs["labels"] = targets["input_ids"]
    return inputs

train_tokens = train_data.map(
    preprocess,
    batched=True,
    remove_columns=train_data.column_names
)

validation_tokens = validation_data.map(
    preprocess,
    batched=True,
    remove_columns=validation_data.column_names
)

collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model
)

train_loader = DataLoader(
    train_tokens,
    batch_size=1,
    shuffle=True,
    collate_fn=collator
)

validation_loader = DataLoader(
    validation_tokens,
    batch_size=1,
    shuffle=False,
    collate_fn=collator
)

print("Training examples:", len(train_tokens))
print("Validation examples:", len(validation_tokens))
````

````python

import torch

model.train()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=2e-5
)

epochs = 1
gradient_accumulation_steps = 4

optimizer.zero_grad()
total_loss = 0.0

for epoch in range(epochs):
    for step, batch in enumerate(train_loader):
        batch = {
            key: value.to(device)
            for key, value in batch.items()
        }

        outputs = model(**batch)
        loss = outputs.loss

        (loss / gradient_accumulation_steps).backward()
        total_loss += loss.item()

        if (
            (step + 1) % gradient_accumulation_steps == 0
            or (step + 1) == len(train_loader)
        ):
            optimizer.step()
            optimizer.zero_grad()

        if (step + 1) % 50 == 0:
            print(
                f"Epoch {epoch + 1}, "
                f"Step {step + 1}/{len(train_loader)}, "
                f"Loss: {loss.item():.4f}"
            )

print("Training finished!")
print("Average training loss:", total_loss / len(train_loader))

model.eval()
````

````python

save_path = "/content/samsum_finetuned"

model.save_pretrained(save_path)
tokenizer.save_pretrained(save_path)

print("Model saved at:", save_path)
````

````python

import torch

model.eval()

# Use the same test examples as before
test_data = dataset["test"].select(range(20))

fine_tuned_summaries = []

for i, example in enumerate(test_data):
    inputs = tokenizer(
        example["dialogue"],
        return_tensors="pt",
        max_length=512,
        truncation=True
    ).to(device)

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=100,
            num_beams=4,
            do_sample=False
        )

    summary = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )

    fine_tuned_summaries.append(summary)

print("Fine-tuned summaries generated:", len(fine_tuned_summaries))
print("\nExample 1 — Human-written summary:")
print(test_data[0]["summary"])

print("\nExample 1 — Fine-tuned summary:")
print(fine_tuned_summaries[0])
````

````python

def generate_image(pipe, prompt, params):
    # Generate images
    img = pipe(prompt, **params).images
    num_images = len(img)

    # Create a row of subplots
    fig, ax = plt.subplots(
        nrows=1,
        ncols=num_images,
        figsize=(6 * num_images, 6)
    )

    # Make ax iterable even when there is only one image
    if num_images == 1:
        ax = [ax]

    for i in range(num_images):
        ax[i].imshow(img[i])
        ax[i].axis("off")

    plt.tight_layout()
    plt.show()
````

````python

import os
import sqlite3
import asyncio

from google.colab import userdata
from aiogram import Bot, Dispatcher
from aiogram.filters import Command
from aiogram.types import Message
from groq import AsyncGroq

# ==========================================
# 1. CONFIGURATION
# ==========================================

TELEGRAM_TOKEN = userdata.get("TELEGRAM_BOT_TOKEN")
GROQ_API_KEY = userdata.get("GROQ_API_KEY")

MODEL = "openai/gpt-oss-20b"

DB_PATH = "/content/drive/MyDrive/telegram_bot_history.db"

SYSTEM_PROMPT = """
You are a friendly and helpful AI assistant.
Answer clearly, accurately, and politely.
Use the conversation history when relevant.
If you are unsure, say so.
"""

bot = Bot(token=TELEGRAM_TOKEN)
dp = Dispatcher()
groq = AsyncGroq(api_key=GROQ_API_KEY)


# ==========================================
# 2. DATABASE SETUP
# ==========================================

def init_db():
    os.makedirs(os.path.dirname(DB_PATH), exist_ok=True)

    with sqlite3.connect(DB_PATH, timeout=30) as conn:
        conn.execute("""
            CREATE TABLE IF NOT EXISTS messages (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                chat_id INTEGER NOT NULL,
                role TEXT NOT NULL,
                content TEXT NOT NULL,
                created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
            )
        """)

        conn.execute("""
            CREATE INDEX IF NOT EXISTS idx_chat_messages
            ON messages(chat_id, id)
        """)

    print("Persistent conversation database is ready!")


def save_message(chat_id, role, content):
    with sqlite3.connect(DB_PATH, timeout=30) as conn:
        conn.execute(
            """
            INSERT INTO messages (chat_id, role, content)
            VALUES (?, ?, ?)
            """,
            (chat_id, role, content)
        )


def load_history(chat_id, limit=12):
    with sqlite3.connect(DB_PATH, timeout=30) as conn:
        rows = conn.execute(
            """
            SELECT role, content
            FROM messages
            WHERE chat_id = ?
            ORDER BY id DESC
            LIMIT ?
            """,
            (chat_id, limit)
        ).fetchall()

    return [
        {"role": role, "content": content}
        for role, content in reversed(rows)
    ]


def clear_history(chat_id):
    with sqlite3.connect(DB_PATH, timeout=30) as conn:
        conn.execute(
            "DELETE FROM messages WHERE chat_id = ?",
            (chat_id,)
        )


# ==========================================
# 3. TELEGRAM COMMANDS
# ==========================================

@dp.message(Command("start"))
async def start_command(message: Message):
    await message.answer(
        "Hello! 👋 Welcome to your Groq AI chatbot.\n\n"
        "Send me any question to start chatting.\n\n"
        "Commands:\n"
        "/start - Start the bot\n"
        "/help - Show help\n"
        "/clear - Clear saved conversation history"
    )


@dp.message(Command("help"))
async def help_command(message: Message):
    await message.answer(
        "🤖 How to use this bot\n\n"
        "• Send a text message to chat with AI.\n"
        "• Your conversation history is saved in Google Drive.\n"
        "• /clear deletes the saved history for this chat.\n"
        "• /help shows this message."
    )


@dp.message(Command("clear"))
async def clear_command(message: Message):
    try:
        clear_history(message.chat.id)
        await message.answer(
            "🧹 Your saved conversation history has been cleared."
        )
    except Exception as e:
        print("Database error while clearing history:", type(e).__name__)
        await message.answer(
            "I couldn't clear your history. Please try again."
        )


# ==========================================
# 4. AI MESSAGE HANDLER
# ==========================================

@dp.message()
async def handle_message(message: Message):
    if not message.text:
        await message.answer(
            "Please send a text message for now."
        )
        return

    user_text = message.text.strip()

    if not user_text:
        return

    chat_id = message.chat.id

    try:
        # Save the user's message first.
        save_message(chat_id, "user", user_text)

        # Load the most recent saved conversation context.
        history = load_history(chat_id, limit=12)

        await bot.send_chat_action(chat_id, "typing")

        response = await groq.chat.completions.create(
            model=MODEL,
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                *history
            ],
            temperature=0.7,
            max_tokens=800
        )

        answer = response.choices[0].message.content

        if not answer:
            answer = (
                "Sorry, I couldn't generate a response. "
                "Please try again."
            )

        # Save the assistant response for future conversations.
        save_message(chat_id, "assistant", answer)

        # Telegram message length is limited.
        for i in range(0, len(answer), 4000):
            await message.answer(answer[i:i + 4000])

        print(f"Replied successfully to chat {chat_id}.")

    except Exception as e:
        # Keep credentials and sensitive exception details private.
        print("Message processing failed:", type(e).__name__)

        await message.answer(
            "Sorry, I couldn't process your message. "
            "Please try again in a moment."
        )


# ==========================================
# 5. STARTUP AND PENDING UPDATES
# ==========================================

async def main():
    init_db()

    # Remove an old webhook if one exists.
    # Do not discard pending Telegram updates.
    webhook_info = await bot.get_webhook_info()

    if webhook_info.url:
        await bot.delete_webhook(drop_pending_updates=False)
        print("Old webhook removed; pending updates preserved.")

    me = await bot.get_me()
    print(f"Bot username: @{me.username}")
    print("Starting polling...")
    print("Persistent history is enabled.")
    print("Keep this Colab cell running while using the bot.")

    try:
        await dp.start_polling(
            bot,
            handle_as_tasks=False,
            close_bot_session=False
        )
    finally:
        await bot.session.close()
        await groq.close()


# Start the bot in Google Colab.
await main()
````

````python

import os
import sqlite3
import asyncio

from google.colab import userdata
from aiogram import Bot, Dispatcher
from aiogram.filters import Command
from aiogram.types import Message
from groq import AsyncGroq

# ==========================================
# 1. CONFIGURATION
# ==========================================

TELEGRAM_TOKEN = userdata.get("TELEGRAM_BOT_TOKEN")
GROQ_API_KEY = userdata.get("GROQ_API_KEY")

MODEL = "openai/gpt-oss-20b"
DB_PATH = "/content/drive/MyDrive/telegram_bot_history.db"

SYSTEM_PROMPT = """
You are a friendly, helpful AI assistant.
Answer clearly, accurately, and politely.
Use conversation history when relevant.
If you are unsure, say so.
"""

bot = Bot(token=TELEGRAM_TOKEN)
dp = Dispatcher()
groq = AsyncGroq(api_key=GROQ_API_KEY)


# ==========================================
# 2. DATABASE
# ==========================================

def init_db():
    os.makedirs(os.path.dirname(DB_PATH), exist_ok=True)

    with sqlite3.connect(DB_PATH, timeout=30) as conn:
        conn.execute("""
            CREATE TABLE IF NOT EXISTS messages (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                chat_id INTEGER NOT NULL,
                role TEXT NOT NULL,
                content TEXT NOT NULL,
                created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
            )
        """)

        conn.execute("""
            CREATE INDEX IF NOT EXISTS idx_chat_messages
            ON messages(chat_id, id)
        """)

    print("Persistent conversation database is ready!")


def load_history(chat_id, limit=12):
    with sqlite3.connect(DB_PATH, timeout=30) as conn:
        rows = conn.execute("""
            SELECT role, content
            FROM messages
            WHERE chat_id = ?
            ORDER BY id DESC
            LIMIT ?
        """, (chat_id, limit)).fetchall()

    return [
        {"role": role, "content": content}
        for role, content in reversed(rows)
    ]


def save_exchange(chat_id, user_text, assistant_text):
    # Save both messages together in one transaction.
    with sqlite3.connect(DB_PATH, timeout=30) as conn:
        conn.execute("BEGIN")
        conn.executemany("""
            INSERT INTO messages (chat_id, role, content)
            VALUES (?, ?, ?)
        """, [
            (chat_id, "user", user_text),
            (chat_id, "assistant", assistant_text),
        ])
        conn.commit()


def clear_history(chat_id):
    with sqlite3.connect(DB_PATH, timeout=30) as conn:
        conn.execute(
            "DELETE FROM messages WHERE chat_id = ?",
            (chat_id,)
        )


# ==========================================
# 3. TELEGRAM COMMANDS
# ==========================================

@dp.message(Command("start"))
async def start_command(message: Message):
    await message.answer(
        "Hello! 👋 I'm your Groq-powered AI assistant.\n\n"
        "Send me a message to chat with AI.\n\n"
        "/start - Start the bot\n"
        "/help - Show commands\n"
        "/clear - Clear saved conversation history"
    )


@dp.message(Command("help"))
async def help_command(message: Message):
    await message.answer(
        "🤖 How to use me\n\n"
        "• Send a text message to chat with AI.\n"
        "• Your completed conversations are saved in Google Drive.\n"
        "• /clear deletes the saved history for this chat.\n"
        "• /help displays these instructions."
    )


@dp.message(Command("clear"))
async def clear_command(message: Message):
    try:
        clear_history(message.chat.id)
        await message.answer(
            "🧹 Your saved conversation history has been cleared."
        )
    except Exception as e:
        print("Clear history failed:", type(e).__name__)
        await message.answer(
            "I couldn't clear your history. Please try again."
        )


# ==========================================
# 4. GROQ REQUEST WITH RETRIES
# ==========================================

async def generate_reply(chat_id, user_text):
    history = load_history(chat_id, limit=12)

    request_messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        *history,
        {"role": "user", "content": user_text},
    ]

    for attempt in range(1, 4):
        try:
            response = await groq.chat.completions.create(
                model=MODEL,
                messages=request_messages,
                temperature=0.7,
                max_tokens=800,
            )

            answer = response.choices[0].message.content

            if not answer:
                raise ValueError("Empty AI response")

            return answer

        except Exception as e:
            status = getattr(e, "status_code", None)

            # Retry likely transient failures, but don't repeatedly
            # retry permanent client errors such as invalid model/key.
            retryable = (
                status is None or status == 408
                or status == 429 or status >= 500
            )

            print(
                f"Groq attempt {attempt} failed: "
                f"{type(e).__name__}"
            )

            if not retryable or attempt == 3:
                return None

            await asyncio.sleep(2 * attempt)

    return None


# ==========================================
# 5. MESSAGE HANDLER
# ==========================================

@dp.message()
async def handle_message(message: Message):
    if not message.text:
        await message.answer("Please send a text message for now.")
        return

    user_text = message.text.strip()
    if not user_text:
        return

    chat_id = message.chat.id

    try:
        await bot.send_chat_action(chat_id, "typing")

        answer = await generate_reply(chat_id, user_text)

        if answer is None:
            await message.answer(
                "Sorry, I couldn't get an AI response after retrying. "
                "Your message was not saved to conversation history. "
                "Please send it again later."
            )
            return

        # Save the user message and AI reply together only after
        # a successful Groq response.
        save_exchange(chat_id, user_text, answer)

    except Exception as e:
        print("Message processing failed:", type(e).__name__)
        await message.answer(
            "Sorry, something went wrong while processing your message."
        )
        return

    try:
        # Split long replies into Telegram-sized chunks.
        for i in range(0, len(answer), 4000):
            await message.answer(answer[i:i + 4000])

        print(f"Saved and replied to chat {chat_id}.")

    except Exception as e:
        print("Telegram delivery failed:", type(e).__name__)
        # The exchange is saved. Don't automatically repeat the AI call.


# ==========================================
# 6. STARTUP AND PENDING UPDATES
# ==========================================

async def main():
    init_db()

    webhook_info = await bot.get_webhook_info()

    if webhook_info.url:
        # Remove webhook without dropping pending messages.
        await bot.delete_webhook(drop_pending_updates=False)
        print("Old webhook removed; pending updates preserved.")

    me = await bot.get_me()
    print(f"Bot username: @{me.username}")
    print("Starting polling...")
    print("Persistent memory enabled.")

    await dp.start_polling(
        bot,
        handle_as_tasks=False,
        close_bot_session=False,
    )


# ==========================================
# 7. RUN
# ==========================================

try:
    await main()
except KeyboardInterrupt:
    print("Bot stopped.")
finally:
    await bot.session.close()
    await groq.close()
````

## Fine tunng SAP project







Fine tuning project


**SupportTicketAI — Fine-Tuned NLP for SAP MDG Support-Ticket Classification**

- Developed an NLP classification prototype using Hugging Face Transformers and DistilBERT to categorize synthetic SAP MDG support tickets into Replication, Change Request, Data Quality, and Access Issue categories.
- Implemented a training, validation, and held-out test workflow with accuracy, macro F1, classification reports, and confusion-matrix evaluation.
- Compared a keyword-based baseline against fine-tuned DistilBERT on an eight-ticket held-out test set. The baseline achieved 87.5% accuracy and 0.867 macro F1, while fine-tuned DistilBERT achieved 75.0% accuracy and 0.733 macro F1.
- Identified replication and access-ticket misclassification as areas for future improvement; documented the limitations of the small synthetic dataset.

**Technologies:** Python, PyTorch, Hugging Face Transformers, DistilBERT, scikit-learn, Google Colab.

````python

from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.25,
    random_state=42,
    stratify=df["label"]
)

train_df = train_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("Training examples:", len(train_df))
print("Testing examples:", len(test_df))

print("\nTraining labels:")
print(train_df["category"].value_counts().sort_index())

print("\nTesting labels:")
print(test_df["category"].value_counts().sort_index())
````

````python

def keyword_baseline(text):
    text = text.lower()

    rules = {
        "Replication": [
            "replication", "target system", "not reaching"
        ],
        "Change_Request": [
            "change request", "workflow", "approval", "submit"
        ],
        "Data_Quality": [
            "missing", "invalid", "duplicate", "validation", "empty"
        ],
        "Access_Issue": [
            "access", "permission", "login", "denied"
        ],
    }

    for category, keywords in rules.items():
        if any(keyword in text for keyword in keywords):
            return category

    return "Unknown"


test_df["baseline_prediction"] = test_df["text"].apply(
    keyword_baseline
)

print(test_df[["text", "category", "baseline_prediction"]].to_string(index=False))
````

````python

from sklearn.metrics import accuracy_score, classification_report

y_true = test_df["category"]
y_pred = test_df["baseline_prediction"]

print("Baseline accuracy:", round(accuracy_score(y_true, y_pred), 3))
print("\nClassification report:")
print(
    classification_report(
        y_true,
        y_pred,
        labels=list(label_names.values()),
        zero_division=0
    )
)
````

````python

import torch
import numpy as np
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

MODEL_NAME = "distilbert/distilbert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

id2label = {
    0: "Replication",
    1: "Change_Request",
    2: "Data_Quality",
    3: "Access_Issue",
}

label2id = {name: idx for idx, name in id2label.items()}

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=4,
    id2label=id2label,
    label2id=label2id
)

print("Model loaded:", MODEL_NAME)
print("Number of classes:", model.config.num_labels)
print("Device:", "GPU" if torch.cuda.is_available() else "CPU")
````

````python

from sklearn.metrics import accuracy_score, classification_report

model.eval()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

test_texts = test_df["text"].tolist()
test_labels = test_df["label"].tolist()

inputs = tokenizer(
    test_texts,
    padding=True,
    truncation=True,
    max_length=128,
    return_tensors="pt"
).to(device)

with torch.no_grad():
    outputs = model(**inputs)
    predictions = torch.argmax(outputs.logits, dim=1).cpu().numpy()

untrained_accuracy = accuracy_score(test_labels, predictions)

print("Untrained model accuracy:", round(untrained_accuracy, 3))
print("\nClassification report:")
print(
    classification_report(
        test_labels,
        predictions,
        labels=[0, 1, 2, 3],
        target_names=[id2label[i] for i in range(4)],
        zero_division=0
    )
)
````

````python

from datasets import Dataset
from transformers import (
    DataCollatorWithPadding,
    TrainingArguments,
    Trainer
)

# Convert training and test data to Hugging Face datasets.
train_dataset = Dataset.from_pandas(
    train_df[["text", "label"]].reset_index(drop=True)
)

test_dataset = Dataset.from_pandas(
    test_df[["text", "label"]].reset_index(drop=True)
)

# Tokenize ticket text.
def tokenize_tickets(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        max_length=128
    )

train_dataset = train_dataset.map(
    tokenize_tickets,
    batched=True
)

test_dataset = test_dataset.map(
    tokenize_tickets,
    batched=True
)

# Pad each batch to the longest ticket in that batch.
data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer
)

print("Training examples:", len(train_dataset))
print("Test examples:", len(test_dataset))
print("Tokenization completed.")
````

````python

import numpy as np
from transformers import TrainingArguments, Trainer

def compute_metrics(eval_pred):
    from sklearn.metrics import accuracy_score, f1_score

    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    return {
        "accuracy": accuracy_score(labels, predictions),
        "macro_f1": f1_score(
            labels,
            predictions,
            average="macro",
            zero_division=0
        )
    }

training_args = TrainingArguments(
    output_dir="./supportticketai_results",
    num_train_epochs=10,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    learning_rate=2e-5,
    weight_decay=0.01,
    eval_strategy="epoch",
    save_strategy="no",
    logging_strategy="epoch",
    report_to="none",
    seed=42,
    data_seed=42
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    data_collator=data_collator,
    processing_class=tokenizer,
    compute_metrics=compute_metrics
)

print("Training configuration ready.")
print("Training examples:", len(train_dataset))
print("Epochs:", training_args.num_train_epochs)
````

````python

training_result = trainer.train()

print("\nTraining completed!")
print("Training metrics:")
print(training_result.metrics)
````

````python

import pandas as pd
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Get predictions from the fine-tuned model.
fine_tuned_output = trainer.predict(test_dataset)
fine_tuned_predictions = np.argmax(
    fine_tuned_output.predictions,
    axis=1
)
true_labels = np.array(test_df["label"])

# Convert keyword baseline predictions to numeric labels.
baseline_label_to_id = {
    category: label
    for label, category in label_names.items()
}

baseline_predictions = np.array([
    baseline_label_to_id.get(pred, -1)
    for pred in test_df["baseline_prediction"]
])

results = pd.DataFrame([
    {
        "Model": "Keyword baseline",
        "Accuracy": accuracy_score(
            true_labels, baseline_predictions
        ),
        "Macro F1": f1_score(
            true_labels, baseline_predictions,
            labels=[0, 1, 2, 3],
            average="macro",
            zero_division=0
        )
    },
    {
        "Model": "Untrained DistilBERT",
        "Accuracy": untrained_accuracy,
        "Macro F1": np.nan
    },
    {
        "Model": "Fine-tuned DistilBERT",
        "Accuracy": accuracy_score(
            true_labels, fine_tuned_predictions
        ),
        "Macro F1": f1_score(
            true_labels, fine_tuned_predictions,
            labels=[0, 1, 2, 3],
            average="macro",
            zero_division=0
        )
    }
])

print(results.to_string(index=False))

print("\nFine-tuned model classification report:")
print(classification_report(
    true_labels,
    fine_tuned_predictions,
    labels=[0, 1, 2, 3],
    target_names=[id2label[i] for i in range(4)],
    zero_division=0
))

print("Fine-tuned model confusion matrix:")
print(confusion_matrix(
    true_labels,
    fine_tuned_predictions,
    labels=[0, 1, 2, 3]
))
````

````python

import torch
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

MODEL_NAME = "distilbert/distilbert-base-uncased"

id2label = {
    0: "Replication",
    1: "Change_Request",
    2: "Data_Quality",
    3: "Access_Issue"
}

label2id = {name: idx for idx, name in id2label.items()}

fresh_tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

fresh_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=4,
    id2label=id2label,
    label2id=label2id
)

print("Fresh pretrained model loaded.")
print("Training rows:", len(train_df))
print("Validation rows:", len(val_df))
print("Reserved final test rows:", len(final_test_df))
print("GPU available:", torch.cuda.is_available())
````

````python

from datasets import Dataset
from transformers import DataCollatorWithPadding

fresh_train_dataset = Dataset.from_pandas(
    train_df[["text", "label"]].reset_index(drop=True)
)

fresh_val_dataset = Dataset.from_pandas(
    val_df[["text", "label"]].reset_index(drop=True)
)

def tokenize_fresh(batch):
    return fresh_tokenizer(
        batch["text"],
        truncation=True,
        max_length=128
    )

fresh_train_dataset = fresh_train_dataset.map(
    tokenize_fresh,
    batched=True
)

fresh_val_dataset = fresh_val_dataset.map(
    tokenize_fresh,
    batched=True
)

fresh_collator = DataCollatorWithPadding(
    tokenizer=fresh_tokenizer
)

print("Training dataset tokenized:", len(fresh_train_dataset))
print("Validation dataset tokenized:", len(fresh_val_dataset))
print("Final test set has not been tokenized or evaluated.")
````

````python

import numpy as np
from sklearn.metrics import accuracy_score, f1_score
from transformers import (
    TrainingArguments,
    Trainer,
    EarlyStoppingCallback
)

def compute_fresh_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    return {
        "accuracy": accuracy_score(labels, predictions),
        "macro_f1": f1_score(
            labels,
            predictions,
            labels=[0, 1, 2, 3],
            average="macro",
            zero_division=0
        )
    }

fresh_training_args = TrainingArguments(
    output_dir="./supportticketai_clean_run",
    num_train_epochs=15,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    learning_rate=2e-5,
    weight_decay=0.01,

    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="eval_macro_f1",
    greater_is_better=True,
    save_total_limit=2,

    logging_strategy="epoch",
    report_to="none",
    seed=42,
    data_seed=42
)

fresh_trainer = Trainer(
    model=fresh_model,
    args=fresh_training_args,
    train_dataset=fresh_train_dataset,
    eval_dataset=fresh_val_dataset,
    data_collator=fresh_collator,
    processing_class=fresh_tokenizer,
    compute_metrics=compute_fresh_metrics,
    callbacks=[
        EarlyStoppingCallback(early_stopping_patience=3)
    ]
)

print("Training configuration ready.")
print("Checkpoint selection: validation macro F1")
print("Final test set will not be used during training.")
````

````python

fresh_training_result = fresh_trainer.train()

print("\nFresh training run completed.")
print("Training metrics:")
print(fresh_training_result.metrics)

print("\nBest checkpoint:")
print(fresh_trainer.state.best_model_checkpoint)

print("\nBest validation macro F1:")
print(fresh_trainer.state.best_metric)
````

````python

import numpy as np
import pandas as pd

from datasets import Dataset
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Prepare the reserved final test set.
final_test_dataset = Dataset.from_pandas(
    final_test_df[["text", "label"]].reset_index(drop=True)
)

def tokenize_final_test(batch):
    return fresh_tokenizer(
        batch["text"],
        truncation=True,
        max_length=128
    )

final_test_dataset = final_test_dataset.map(
    tokenize_final_test,
    batched=True
)

# Predict using the selected best checkpoint.
final_output = fresh_trainer.predict(final_test_dataset)

final_predictions = np.argmax(
    final_output.predictions,
    axis=1
)

final_labels = np.array(final_test_df["label"])

# Evaluate the keyword baseline on exactly the same test tickets.
final_test_df = final_test_df.copy()

final_test_df["baseline_prediction"] = (
    final_test_df["text"].apply(keyword_baseline)
)

category_to_id = {
    name: label for label, name in label_names.items()
}

baseline_final_predictions = (
    final_test_df["baseline_prediction"]
    .map(category_to_id)
    .fillna(-1)
    .astype(int)
    .to_numpy()
)

# Calculate final metrics.
print("FINAL TEST RESULTS")
print("------------------")

print("Fine-tuned DistilBERT accuracy:",
      round(accuracy_score(final_labels, final_predictions), 3))

print("Fine-tuned DistilBERT macro F1:",
      round(f1_score(
          final_labels,
          final_predictions,
          labels=[0, 1, 2, 3],
          average="macro",
          zero_division=0
      ), 3))

print("\nKeyword baseline accuracy:",
      round(accuracy_score(
          final_labels,
          baseline_final_predictions
      ), 3))

print("Keyword baseline macro F1:",
      round(f1_score(
          final_labels,
          baseline_final_predictions,
          labels=[0, 1, 2, 3],
          average="macro",
          zero_division=0
      ), 3))

print("\nFine-tuned model classification report:")
print(classification_report(
    final_labels,
    final_predictions,
    labels=[0, 1, 2, 3],
    target_names=[id2label[i] for i in range(4)],
    zero_division=0
))

print("Fine-tuned model confusion matrix:")
print(confusion_matrix(
    final_labels,
    final_predictions,
    labels=[0, 1, 2, 3]
))
````

````python
def translate_audio(audio_filename, target_language="German"):
    # Step 1: Transcribe the audio
    with open(audio_filename, "rb") as audio_file:
        transcription = client.audio.transcriptions.create(
            file=(audio_filename, audio_file.read()),
            model="whisper-large-v3-turbo",
            response_format="text"
        )

    transcript = transcription.strip()

    # Step 2: Translate the transcript
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "system",
                "content": (
                    f"Translate the text into {target_language}. "
                    "Preserve its meaning and return only the translation."
                )
            },
            {"role": "user", "content": transcript}
        ],
        temperature=0
    )

    translated_text = response.choices[0].message.content.strip()

    return transcript, translated_text


# Example usage
transcript, translation = translate_audio(
    audio_filename,
    target_language="German"
)

print("Original transcription:\n", transcript)
print("\nGerman translation:\n", translation)
````

````python

# # Install if needed (run once)
# !pip install -q -U groq gradio

# import os
# import tempfile
# from datetime import datetime

# import gradio as gr
# from groq import Groq
# from google.colab import userdata

# # Connect to Groq using Colab Secrets
# client = Groq(api_key=userdata.get("GROQ_API_KEY"))


# def translate_audio(audio_path, target_language):
#     if not audio_path:
#         raise gr.Error("Please upload an audio file.")

#     if not target_language:
#         raise gr.Error("Please select a target language.")

#     try:
#         # 1. Transcribe audio
#         with open(audio_path, "rb") as audio_file:
#             result = client.audio.transcriptions.create(
#                 file=(os.path.basename(audio_path), audio_file),
#                 model="whisper-large-v3-turbo",
#                 response_format="text"
#             )

#         transcript = result.strip()

#         if not transcript:
#             raise gr.Error("No speech was detected.")

#         # 2. Translate text
#         response = client.chat.completions.create(
#             model="openai/gpt-oss-20b",
#             messages=[
#                 {
#                     "role": "system",
#                     "content": (
#                         f"Translate the text into {target_language}. "
#                         "Preserve its meaning and tone. "
#                         "Return only the translation."
#                     )
#                 },
#                 {"role": "user", "content": transcript}
#             ],
#             temperature=0
#         )

#         translation = (
#             response.choices[0].message.content or ""
#         ).strip()

#         return transcript, translation

#     except gr.Error:
#         raise
#     except Exception as e:
#         print("Translation error:", repr(e))
#         raise gr.Error(
#             "Translation failed. Check the Colab output for details."
#         )


# def save_translation_txt(transcript, translation, target_language):
#     if not transcript or not translation:
#         raise gr.Error(
#             "Please translate an audio file before downloading."
#         )

#     content = (
#         "AUDIO TRANSLATION RESULTS\n"
#         "=========================\n\n"
#         "ORIGINAL TRANSCRIPTION\n"
#         "----------------------\n"
#         f"{transcript}\n\n"
#         f"TRANSLATION ({target_language})\n"
#         "----------------------\n"
#         f"{translation}\n"
#     )

#     # Create a real temporary file and return its path to Gradio
#     with tempfile.NamedTemporaryFile(
#         mode="w",
#         encoding="utf-8",
#         suffix=".txt",
#         prefix="audio_translation_",
#         delete=False
#     ) as txt_file:
#         txt_file.write(content)
#         filepath = txt_file.name

#     return filepath


# with gr.Blocks(title="Groq Audio Translator") as demo:
#     gr.Markdown("# 🎙️ Groq Audio Translator")
#     gr.Markdown(
#         "Upload audio, translate the speech, and download both texts."
#     )

#     audio = gr.Audio(
#         label="Upload Audio",
#         type="filepath"
#     )

#     language = gr.Dropdown(
#         choices=[
#             "German", "Hindi", "English", "French",
#             "Spanish", "Arabic", "Urdu", "Japanese"
#         ],
#         value="German",
#         label="Translate Into"
#     )

#     translate_button = gr.Button(
#         "Translate Audio",
#         variant="primary"
#     )

#     original = gr.Textbox(
#         label="Original Transcription",
#         lines=6
#     )

#     translated = gr.Textbox(
#         label="Translated Text",
#         lines=6
#     )

#     download_button = gr.Button("📥 Download Results as TXT")

#     download_file = gr.File(
#         label="Your TXT file",
#         interactive=False
#     )

#     translate_button.click(
#         fn=translate_audio,
#         inputs=[audio, language],
#         outputs=[original, translated]
#     )

#     download_button.click(
#         fn=save_translation_txt,
#         inputs=[original, translated, language],
#         outputs=[download_file]
#     )

# provide me updated code
# demo.launch(share=True)


!pip install -q -U groq gradio

import os
import tempfile
from datetime import datetime

import gradio as gr
from groq import Groq
from google.colab import userdata, files

# Connect to Groq using Colab Secrets
client = Groq(api_key=userdata.get("GROQ_API_KEY"))

# Store the latest generated TXT file path
last_txt_path = None


def translate_audio(audio_path, target_language):
    if not audio_path:
        raise gr.Error("Please upload an audio file.")

    if not target_language:
        raise gr.Error("Please select a target language.")

    try:
        # 1. Transcribe audio
        with open(audio_path, "rb") as audio_file:
            result = client.audio.transcriptions.create(
                file=(os.path.basename(audio_path), audio_file),
                model="whisper-large-v3-turbo",
                response_format="text"
            )

        transcript = result.strip()

        if not transcript:
            raise gr.Error("No speech was detected.")

        # 2. Translate text
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[
                {
                    "role": "system",
                    "content": (
                        f"Translate the text into {target_language}. "
                        "Preserve its meaning and tone. "
                        "Return only the translation."
                    )
                },
                {"role": "user", "content": transcript}
            ],
            temperature=0
        )

        translation = (
            response.choices[0].message.content or ""
        ).strip()

        return transcript, translation

    except gr.Error:
        raise
    except Exception as e:
        print("Translation error:", repr(e))
        raise gr.Error(
            "Translation failed. Check the Colab output for details."
        )


def save_translation_txt(transcript, translation, target_language):
    global last_txt_path

    if not transcript or not translation:
        raise gr.Error(
            "Please translate audio before downloading."
        )

    content = (
        "AUDIO TRANSLATION RESULTS\n"
        "=========================\n\n"
        "ORIGINAL TRANSCRIPTION\n"
        "----------------------\n"
        f"{transcript}\n\n"
        f"TRANSLATION ({target_language})\n"
        "----------------------\n"
        f"{translation}\n"
    )

    filename = (
        f"audio_translation_{datetime.now():%Y%m%d_%H%M%S}.txt"
    )
    last_txt_path = os.path.join(tempfile.gettempdir(), filename)

    with open(last_txt_path, "w", encoding="utf-8") as txt_file:
        txt_file.write(content)

    # Return the actual file path to Gradio
    return last_txt_path


# Build the web interface
with gr.Blocks(title="Groq Audio Translator") as demo:
    gr.Markdown("# 🎙️ Groq Audio Translator")
    gr.Markdown(
        "Upload audio, translate the speech, and download both texts."
    )

    audio = gr.Audio(
        label="Upload Audio",
        type="filepath"
    )

    language = gr.Dropdown(
        choices=[
            "German", "Hindi", "English", "French",
            "Spanish", "Arabic", "Urdu", "Japanese"
        ],
        value="German",
        label="Translate Into"
    )

    translate_button = gr.Button(
        "Translate Audio",
        variant="primary"
    )

    original = gr.Textbox(
        label="Original Transcription",
        lines=6
    )

    translated = gr.Textbox(
        label="Translated Text",
        lines=6
    )

    download_button = gr.Button("📥 Prepare TXT Download")

    download_file = gr.File(
        label="TXT File",
        interactive=False
    )

    translate_button.click(
        fn=translate_audio,
        inputs=[audio, language],
        outputs=[original, translated]
    )

    download_button.click(
        fn=save_translation_txt,
        inputs=[original, translated, language],
        outputs=[download_file]
    )


demo.queue()
demo.launch(share=True)

````

````python
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import tiktoken

print("All imports successful!")
````

````python
documents = [
    Document(
        page_content="ChromaDB is a vector database used to store embeddings and perform similarity search.",
        metadata={"source": "chromadb.txt"}
    ),
    Document(
        page_content="LangChain helps developers build applications using language models, prompts, retrievers, and document loaders.",
        metadata={"source": "langchain.txt"}
    ),
    Document(
        page_content="Groq provides fast inference for supported large language models through its API.",
        metadata={"source": "groq.txt"}
    ),
    Document(
        page_content="Tiktoken is a tokenizer library used to estimate token counts for supported encodings.",
        metadata={"source": "tiktoken.txt"}
    )
]

splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=30
)

chunks = splitter.split_documents(documents)

encoding = tiktoken.get_encoding("cl100k_base")

for i, chunk in enumerate(chunks, 1):
    count = len(encoding.encode(chunk.page_content))
    print(f"Chunk {i}: {count} tokens | Source: {chunk.metadata['source']}")
````

````python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    collection_name="groq_chroma_demo"
)

print("ChromaDB vector store created!")
````

````python
import os
from getpass import getpass
from langchain_groq import ChatGroq

# Re-enter your key if the Colab runtime restarted
if not os.environ.get("GROQ_API_KEY"):
    os.environ["GROQ_API_KEY"] = getpass("Enter your Groq API key: ")

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

response = llm.invoke(
    "Explain vector databases in one sentence."
)

print(response.content)
````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template("""
You are a helpful AI assistant. Answer using the retrieved context only.
If the context does not contain the answer, say you don't know based on these documents.

Context:
{context}

Question:
{question}

Answer:
""")

chain = prompt | llm | StrOutputParser()

def ask_question(question):
    docs = vectorstore.similarity_search(question, k=3)

    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    answer = chain.invoke({
        "context": context,
        "question": question
    })

    sources = sorted({
        doc.metadata.get("source", "unknown")
        for doc in docs
    })

    return answer, sources

answer, sources = ask_question(
    "What is ChromaDB used for?"
)

print("Answer:", answer)
print("Sources:", sources)
````

````python
import gradio as gr

def chat(question):
    if not question.strip():
        return "Please enter a question.", ""

    answer, sources = ask_question(question)

    return answer, ", ".join(sources)

demo = gr.Interface(
    fn=chat,
    inputs=gr.Textbox(
        lines=2,
        label="Ask a question",
        placeholder="What is ChromaDB used for?"
    ),
    outputs=[
        gr.Textbox(label="AI Answer"),
        gr.Textbox(label="Retrieved Sources")
    ],
    title="ChromaDB + LangChain + Groq RAG Demo",
    description="Ask questions about the sample documents stored in ChromaDB."
)

demo.launch(share=True)
````

````python
import gradio as gr
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

chat_prompt = ChatPromptTemplate.from_template("""
You are a helpful PDF assistant.

Use the retrieved document context and conversation history to answer.
If the document does not support an answer, say so clearly.
Treat the conversation history as context, not as a source of document facts.

Conversation history:
{history}

Retrieved document context:
{context}

Current question:
{question}

Answer:
""")

chat_chain = chat_prompt | llm | StrOutputParser()


def pdf_chat_with_history(message, history):
    if not message.strip():
        return "Please enter a question."

    # Rewrite follow-up questions using conversation history
    history_text = "\n".join(
        f"{'User' if item['role'] == 'user' else 'Assistant'}: {item['content']}"
        for item in history[-8:]
    )

    search_query = message

    if history_text:
        rewrite_prompt = ChatPromptTemplate.from_template("""
Rewrite the latest question as a standalone search query using
the conversation history where helpful. Return only the query.

History:
{history}

Latest question:
{question}
""")

        search_chain = rewrite_prompt | llm | StrOutputParser()
        search_query = search_chain.invoke({
            "history": history_text,
            "question": message
        }).strip()

    docs = pdf_vectorstore.similarity_search(search_query, k=4)

    context = "\n\n".join(
        f"[Page {doc.metadata.get('page', 0) + 1}] {doc.page_content}"
        for doc in docs
    )

    answer = chat_chain.invoke({
        "history": history_text or "No previous conversation.",
        "context": context or "No relevant text was retrieved.",
        "question": message
    })

    sources = sorted({
        str(doc.metadata.get("page", 0) + 1)
        for doc in docs
    })

    if sources:
        answer += "\n\n**Source pages:** " + ", ".join(sources)

    return answer


demo = gr.ChatInterface(
    fn=pdf_chat_with_history,
    title="Chat with Your PDF",
    description="Ask questions and follow up naturally. Powered by ChromaDB, LangChain, Groq and tiktoken."
)

demo.launch(share=True)
````

````python

import gradio as gr
import tiktoken
from langchain_core.prompts import ChatPromptTemplate

encoding = tiktoken.get_encoding("cl100k_base")

rag_prompt = ChatPromptTemplate.from_template("""
You are a helpful PDF question-answering assistant.

Rules:
- Answer using the retrieved PDF excerpts.
- Cite supporting passages using their page labels, such as [Page 2].
- If the answer is not supported by the excerpts, say so.
- Do not invent citations.

Conversation history:
{history}

Retrieved PDF excerpts:
{context}

Question:
{question}

Answer:
""")

def pdf_chat_with_citations(message, history):
    if not message.strip():
        return "Please enter a question."

    # Retrieve relevant PDF chunks
    docs = pdf_vectorstore.similarity_search(message, k=4)

    # Build context with page citations
    context_parts = []
    for i, doc in enumerate(docs, 1):
        page = doc.metadata.get("page")
        page_label = f"Page {int(page) + 1}" if page is not None else "Unknown page"
        context_parts.append(
            f"[{page_label}]\n{doc.page_content}"
        )

    context = "\n\n".join(context_parts)
    if not context:
        context = "No relevant passages were retrieved."

    # Include recent conversation history
    history_lines = []
    for item in history[-6:]:
        if isinstance(item, dict):
            role = item.get("role", "user")
            content = item.get("content", "")
            if isinstance(content, str):
                history_lines.append(f"{role}: {content}")

    history_text = "\n".join(history_lines) or "No previous conversation."

    # Prepare the prompt
    messages = rag_prompt.format_messages(
        history=history_text,
        context=context,
        question=message
    )

    prompt_text = "\n".join(
        str(m.content) for m in messages
    )
    estimated_input_tokens = len(encoding.encode(prompt_text))

    # Call Groq and retain usage metadata
    response = llm.invoke(messages)
    answer = response.content
    if not isinstance(answer, str):
        answer = str(answer)

    # List the source pages actually retrieved
    source_pages = sorted({
        int(doc.metadata["page"]) + 1
        for doc in docs
        if doc.metadata.get("page") is not None
    })

    citations = ", ".join(f"[Page {p}]" for p in source_pages)
    if citations:
        answer += f"\n\n**Retrieved source pages:** {citations}"

    # Estimate output tokens
    estimated_output_tokens = len(encoding.encode(answer))

    # Read actual usage when returned by LangChain/Groq
    usage = getattr(response, "usage_metadata", None) or {}
    actual_input = usage.get("input_tokens")
    actual_output = usage.get("output_tokens")

    if actual_input is None or actual_output is None:
        metadata = getattr(response, "response_metadata", {}) or {}
        token_usage = metadata.get("token_usage") or metadata.get("usage") or {}
        actual_input = token_usage.get("prompt_tokens", actual_input)
        actual_output = token_usage.get("completion_tokens", actual_output)

    usage_report = (
        "\n\n---\n"
        "**Token usage**\n"
        f"- Estimated input tokens: {estimated_input_tokens}\n"
        f"- Estimated output tokens: {estimated_output_tokens}\n"
        f"- Groq-reported input tokens: {actual_input if actual_input is not None else 'Not provided'}\n"
        f"- Groq-reported output tokens: {actual_output if actual_output is not None else 'Not provided'}"
    )

    return answer + usage_report


demo = gr.ChatInterface(
    fn=pdf_chat_with_citations,
    title="PDF RAG Chatbot — Sources & Token Usage",
    description="Ask questions about your PDF and inspect retrieved pages and token counts."
)

demo.launch(share=True)
````

````python

import os
import gradio as gr
import tiktoken

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Token counter
encoding = tiktoken.get_encoding("cl100k_base")

# Split uploaded PDFs into searchable chunks
pdf_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

# RAG prompt
multi_pdf_prompt = ChatPromptTemplate.from_template("""
You are a helpful document assistant.

Answer the question using the supplied document excerpts.
Cite evidence using the labels [Source: filename, Page N].
If the documents do not contain the answer, say so.
Do not invent facts or citations.

Document excerpts:
{context}

Question:
{question}

Answer:
""")

multi_pdf_chain = multi_pdf_prompt | llm | StrOutputParser()


def index_multiple_pdfs(files):
    if not files:
        return "Please upload one or more PDF files."

    total_pages = 0
    total_chunks = 0
    indexed_files = []
    errors = []

    for path in files:
        try:
            loader = PyPDFLoader(path)
            pages = loader.load()

            if not pages:
                errors.append(f"{os.path.basename(path)}: no readable text")
                continue

            filename = os.path.basename(path)

            # Preserve original page numbers and use clean filenames
            for page in pages:
                page.metadata["source"] = filename

            chunks = pdf_splitter.split_documents(pages)

            if not chunks:
                errors.append(
                    f"{filename}: no extractable text; OCR may be needed"
                )
                continue

            # Append new chunks to your existing ChromaDB collection
            pdf_vectorstore.add_documents(chunks)

            total_pages += len(pages)
            total_chunks += len(chunks)
            indexed_files.append(filename)

        except Exception as e:
            errors.append(f"{os.path.basename(path)}: {str(e)[:250]}")

    report = (
        f"Indexed files: {len(indexed_files)}\n"
        f"Pages processed: {total_pages}\n"
        f"Chunks added: {total_chunks}\n"
        f"Files: {', '.join(indexed_files) or 'None'}"
    )

    if errors:
        report += "\n\nIssues:\n" + "\n".join(errors)

    return report


def answer_multiple_pdfs(question):
    if not question or not question.strip():
        return "Please enter a question."

    # Search across all documents in the collection
    docs = pdf_vectorstore.similarity_search(question, k=5)

    if not docs:
        return "No relevant document passages were found."

    # Give every excerpt a filename and page label
    context_parts = []

    for doc in docs:
        filename = doc.metadata.get("source", "Unknown file")
        page_number = doc.metadata.get("page")

        page_label = (
            int(page_number) + 1
            if page_number is not None
            else "Unknown"
        )

        context_parts.append(
            f"[Source: {filename}, Page {page_label}]\n"
            f"{doc.page_content}"
        )

    context = "\n\n".join(context_parts)

    # Generate the answer
    answer = multi_pdf_chain.invoke({
        "context": context,
        "question": question
    })

    # Show the actual retrieved sources
    sources = sorted({
        (
            doc.metadata.get("source", "Unknown file"),
            int(doc.metadata["page"]) + 1
        )
        for doc in docs
        if doc.metadata.get("page") is not None
    })

    source_text = "\n".join(
        f"- {filename}, page {page}"
        for filename, page in sources
    )

    # Token counts are estimates, not necessarily Groq's exact counts
    input_estimate = len(
        encoding.encode(context + "\n" + question)
    )
    output_estimate = len(encoding.encode(answer))

    return (
        f"{answer}\n\n"
        f"### Retrieved sources\n{source_text}\n\n"
        f"### Estimated token counts\n"
        f"- Retrieved context + question: {input_estimate}\n"
        f"- Answer: {output_estimate}"
    )


# Gradio application
with gr.Blocks(title="Multi-PDF RAG Assistant") as multi_pdf_demo:
    gr.Markdown("# Multi-PDF RAG Assistant")
    gr.Markdown(
        "Upload multiple PDFs, index them into ChromaDB, "
        "and ask questions across all indexed documents."
    )

    with gr.Row():
        pdf_upload = gr.File(
            label="Upload PDF files",
            file_count="multiple",
            file_types=[".pdf"],
            type="filepath"
        )

    index_button = gr.Button("Index uploaded PDFs", variant="primary")
    indexing_status = gr.Textbox(
        label="Indexing status",
        lines=5,
        interactive=False
    )

    index_button.click(
        fn=index_multiple_pdfs,
        inputs=[pdf_upload],
        outputs=[indexing_status]
    )

    gr.Markdown("## Ask questions across your documents")

    question_box = gr.Textbox(
        label="Your question",
        placeholder="Compare the conclusions of the uploaded PDFs...",
        lines=2
    )

    ask_button = gr.Button("Ask", variant="primary")
    answer_box = gr.Markdown(label="Answer")

    ask_button.click(
        fn=answer_multiple_pdfs,
        inputs=[question_box],
        outputs=[answer_box]
    )

    question_box.submit(
        fn=answer_multiple_pdfs,
        inputs=[question_box],
        outputs=[answer_box]
    )

multi_pdf_demo.launch(share=True)
````

````python
def retrieve_relevant_chunks(query, k=5, max_distance=1.0):
    results = pdf_vectorstore.similarity_search_with_score(
        query,
        k=k
    )

    relevant_docs = [
        (doc, distance)
        for doc, distance in results
        if float(distance) <= max_distance
    ]

    if not relevant_docs:
        print("No chunks passed the distance threshold.")
        print("Try adjusting max_distance after inspecting results.")
        return []

    for doc, distance in relevant_docs:
        print(
            f"\nSource: {doc.metadata.get('source', 'Unknown')}"
            f"\nPage: {doc.metadata.get('page', 'Unknown')}"
            f"\nDistance: {float(distance):.4f}"
            f"\n{doc.page_content[:350]}"
        )

    return [doc for doc, _ in relevant_docs]


relevant_docs = retrieve_relevant_chunks(
    "What are the main findings?",
    k=5,
    max_distance=1.0
)
````

````python
from langchain_community.retrievers import BM25Retriever
from langchain_core.documents import Document

# Read documents already stored in ChromaDB
stored = pdf_vectorstore.get(include=["documents", "metadatas"])

all_docs = [
    Document(
        page_content=text,
        metadata=metadata or {}
    )
    for text, metadata in zip(
        stored["documents"],
        stored["metadatas"]
    )
    if text and text.strip()
]

if not all_docs:
    raise ValueError("No documents found in ChromaDB. Index your PDFs first.")

bm25_retriever = BM25Retriever.from_documents(
    all_docs,
    k=5
)

print("Documents indexed for keyword search:", len(all_docs))
````

````python
def hybrid_search(query, k=5, candidate_k=10, rrf_constant=60):
    # 1. Semantic search using existing ChromaDB
    semantic_results = pdf_vectorstore.similarity_search(
        query,
        k=candidate_k
    )

    # 2. Keyword search using BM25
    keyword_results = bm25_retriever.invoke(query)[:candidate_k]

    # 3. Combine rankings with Reciprocal Rank Fusion
    scores = {}
    documents_by_key = {}

    def document_key(doc):
        return (
            doc.metadata.get("source", ""),
            doc.metadata.get("page", -1),
            doc.page_content
        )

    for result_list in [semantic_results, keyword_results]:
        seen_in_list = set()

        for rank, doc in enumerate(result_list, start=1):
            key = document_key(doc)

            if key in seen_in_list:
                continue

            seen_in_list.add(key)
            documents_by_key[key] = doc

            scores[key] = scores.get(key, 0) + (
                1 / (rrf_constant + rank)
            )

    ranked_keys = sorted(
        scores,
        key=lambda key: scores[key],
        reverse=True
    )

    return [
        (documents_by_key[key], scores[key])
        for key in ranked_keys[:k]
    ]
````

````python
query = "What are the main findings and limitations?"

results = hybrid_search(query, k=5)

for rank, (doc, score) in enumerate(results, start=1):
    print(f"\n--- Hybrid result {rank} ---")
    print("Source:", doc.metadata.get("source", "Unknown"))
    print("Page:", (
        int(doc.metadata["page"]) + 1
        if doc.metadata.get("page") is not None
        else "Unknown"
    ))
    print("RRF score:", round(score, 5))
    print("Text:", doc.page_content[:400])
````

````python

import os
import gradio as gr
import tiktoken

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

encoding = tiktoken.get_encoding("cl100k_base")

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)


def index_pdfs(files):
    if not files:
        return "Please upload one or more PDF files."

    total_chunks = 0
    indexed_files = []
    errors = []

    for file in files:
        path = file if isinstance(file, str) else file.name
        filename = os.path.basename(path)

        try:
            pages = PyPDFLoader(path).load()

            for page in pages:
                page.metadata["source"] = filename

            chunks = splitter.split_documents(pages)

            if not chunks:
                errors.append(f"{filename}: no extractable text")
                continue

            # Add documents to the existing ChromaDB collection
            pdf_vectorstore.add_documents(chunks)

            total_chunks += len(chunks)
            indexed_files.append(filename)

        except Exception as e:
            errors.append(f"{filename}: {str(e)[:200]}")

    # Refresh BM25 using all documents currently in ChromaDB
    stored = pdf_vectorstore.get(
        include=["documents", "metadatas"]
    )

    from langchain_core.documents import Document
    from langchain_community.retrievers import BM25Retriever

    all_docs = [
        Document(page_content=text, metadata=metadata or {})
        for text, metadata in zip(
            stored["documents"],
            stored["metadatas"]
        )
        if text and text.strip()
    ]

    global bm25_retriever
    if all_docs:
        bm25_retriever = BM25Retriever.from_documents(
            all_docs,
            k=10
        )

    report = (
        f"Successfully indexed files: {len(indexed_files)}\n"
        f"Chunks added: {total_chunks}\n"
        f"Files: {', '.join(indexed_files) or 'None'}"
    )

    if errors:
        report += "\n\nIssues:\n" + "\n".join(errors)

    return report


def answer_multiple_pdfs(question):
    if not question or not question.strip():
        return "Please enter a question."

    # Hybrid retrieval: ChromaDB + BM25
    hybrid_results = hybrid_search(question, k=5)
    docs = [doc for doc, score in hybrid_results]

    if not docs:
        return "No relevant passages found. Please index your PDFs first."

    context_parts = []
    sources = set()

    for doc in docs:
        filename = doc.metadata.get("source", "Unknown file")
        page = doc.metadata.get("page")
        page_number = int(page) + 1 if page is not None else None

        page_label = (
            f"Page {page_number}"
            if page_number is not None
            else "Page unknown"
        )

        context_parts.append(
            f"[Source: {filename}, {page_label}]\n"
            f"{doc.page_content}"
        )

        sources.add(
            f"{filename}, {page_label}"
        )

    context = "\n\n".join(context_parts)

    prompt = f"""
You are a helpful PDF assistant.
Answer using the retrieved excerpts only.
Cite the filename and page number for supporting evidence.
If the context is insufficient, say so. Do not invent citations.

Context:
{context}

Question:
{question}

Answer:
"""

    response = llm.invoke(prompt)
    answer = response.content

    if not isinstance(answer, str):
        answer = str(answer)

    source_text = "\n".join(
        f"- {source}" for source in sorted(sources)
    )

    input_tokens = len(
        encoding.encode(context + "\n" + question)
    )
    output_tokens = len(encoding.encode(answer))

    return (
        f"{answer}\n\n"
        f"### Retrieved sources\n{source_text}\n\n"
        f"### Estimated token counts\n"
        f"- Context + question: {input_tokens}\n"
        f"- Answer: {output_tokens}"
    )


# Gradio interface
with gr.Blocks(title="Hybrid Search PDF Chatbot") as demo:
    gr.Markdown("# 📚 Multi-PDF RAG Chatbot")
    gr.Markdown(
        "Powered by ChromaDB, BM25, LangChain, Groq and tiktoken."
    )

    pdf_files = gr.File(
        label="Upload multiple PDF files",
        file_count="multiple",
        file_types=[".pdf"],
        type="filepath"
    )

    index_button = gr.Button(
        "Index PDFs",
        variant="primary"
    )

    index_status = gr.Textbox(
        label="Indexing status",
        lines=5,
        interactive=False
    )

    index_button.click(
        fn=index_pdfs,
        inputs=pdf_files,
        outputs=index_status
    )

    gr.Markdown("## Ask questions across your PDFs")

    question = gr.Textbox(
        label="Your question",
        placeholder="Compare the main findings across my PDFs...",
        lines=2
    )

    ask_button = gr.Button("Search and Answer", variant="primary")
    answer_output = gr.Markdown()

    ask_button.click(
        fn=answer_multiple_pdfs,
        inputs=question,
        outputs=answer_output
    )

    question.submit(
        fn=answer_multiple_pdfs,
        inputs=question,
        outputs=answer_output
    )

demo.launch(share=True)
````

````python
import weaviate
from weaviate.classes.config import Configure, Property, DataType

COLLECTION_NAME = "PDFChunk"

# Delete the old incorrectly configured collection, if it exists
if client.collections.exists(COLLECTION_NAME):
    client.collections.delete(COLLECTION_NAME)
    print("Deleted old PDFChunk collection.")

# Create a collection that accepts vectors from Hugging Face
client.collections.create(
    name=COLLECTION_NAME,
    vector_config=Configure.Vectors.self_provided(),
    properties=[
        Property(name="text", data_type=DataType.TEXT),
        Property(name="source", data_type=DataType.TEXT),
        Property(name="page", data_type=DataType.INT),
    ],
)

print("PDFChunk collection recreated with vector indexing enabled!")
````

````python
from google.colab import files
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
import os

uploaded = files.upload()

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

chunks = []

for filename in uploaded.keys():
    pages = PyPDFLoader(filename).load()

    for page in pages:
        page.metadata["source"] = os.path.basename(filename)
        page.metadata["page"] = int(page.metadata.get("page", 0))

    chunks.extend(splitter.split_documents(pages))

if chunks:
    ids = vectorstore.add_documents(chunks)
    print(f"Successfully indexed {len(ids)} chunks.")
else:
    print("No readable PDF text was extracted.")
````

````python
import wikipedia

def search_wikipedia(query: str) -> str:
    """Search Wikipedia for information about a topic."""
    try:
        return wikipedia.summary(
            query,
            sentences=5,
            auto_suggest=True,
            redirect=True
        )
    except wikipedia.exceptions.DisambiguationError as e:
        return "The topic is ambiguous. Possible matches: " + ", ".join(e.options[:5])
    except wikipedia.exceptions.PageError:
        return "No matching Wikipedia page was found."
    except Exception as e:
        return f"Wikipedia search failed: {type(e).__name__}: {e}"

print(search_wikipedia("Cricket World Cup"))
````

````python
from langchain_core.tools import tool

@tool
def cricket_wikipedia_search(query: str) -> str:
    """Search Wikipedia for factual information about a topic."""
    return search_wikipedia(query)

llm_with_tools = llm.bind_tools([cricket_wikipedia_search])

response = llm_with_tools.invoke(
    "Find information about the first Cricket World Cup."
)

print("Tool calls:", response.tool_calls)
print("Response:", response.content)
````

````python
from langchain_core.messages import HumanMessage, ToolMessage

llm_with_tools = llm.bind_tools([cricket_wikipedia_search])

messages = [
    HumanMessage(
        content="Find information about the first Cricket World Cup."
    )
]

response = llm_with_tools.invoke(messages)
messages.append(response)

for tool_call in response.tool_calls:
    if tool_call["name"] == "cricket_wikipedia_search":
        result = cricket_wikipedia_search.invoke(tool_call["args"])

        messages.append(
            ToolMessage(
                content=result,
                tool_call_id=tool_call["id"]
            )
        )

if response.tool_calls:
    final_response = llm_with_tools.invoke(messages)
    print(final_response.content)
else:
    print(response.content)
````

````python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import InMemoryChatMessageHistory

# Store separate conversation histories by session ID
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

# Create a prompt that includes conversation history
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Use conversation history when relevant."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}")
])

# Build the chain
chain = prompt | llm

# Automatically manage message history
chatbot = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="history"
)

print("Chatbot with managed memory is ready!")
````

````python
from langchain_core.prompts import ChatPromptTemplate

rag_prompt = ChatPromptTemplate.from_template("""
Answer the question using only the context below.
If the context does not contain the answer, say you do not have
enough information to answer.

Context:
{context}

Question:
{question}
""")

def answer_question(question):
    docs = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    response = (rag_prompt | llm).invoke({
        "context": context,
        "question": question
    })

    return response.content

print(answer_question(
    "What are the key features of the Llama 2 model?"
))
````

````python
def answer_question_with_sources(question):
    docs = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    response = (rag_prompt | llm).invoke({
        "context": context,
        "question": question
    })

    sources = list(dict.fromkeys(
        doc.metadata.get("source", "Unknown")
        for doc in docs
    ))

    return {
        "answer": response.content,
        "sources": sources
    }


result = answer_question_with_sources(
    "What are the key features of the MPT-7B language model?"
)

print("ANSWER:\n")
print(result["answer"])

print("\nSOURCES:\n")
for source in result["sources"]:
    print(source)
````

````python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# Step 1: Generate a restaurant name
name_prompt = PromptTemplate.from_template(
    "Suggest one fancy, memorable restaurant name for {cuisine} cuisine."
)

name_chain = name_prompt | llm | StrOutputParser()

# Step 2: Generate menu items using the restaurant name
menu_prompt = PromptTemplate.from_template(
    "Create 5 menu items for the restaurant named {restaurant_name}. "
    "The restaurant serves {cuisine} cuisine."
)

menu_chain = menu_prompt | llm | StrOutputParser()

# Connect both steps into one sequential chain
sequential_chain = (
    RunnablePassthrough.assign(restaurant_name=name_chain)
    .assign(menu_items=menu_chain)
)

# Run the complete chain
result = sequential_chain.invoke({"cuisine": "Indian"})

print("Cuisine:", result["cuisine"])
print("\nRestaurant name:", result["restaurant_name"])
print("\nMenu items:\n", result["menu_items"])
````

````python
from langchain_core.tools import tool

@tool
def multiply_numbers(a: float, b: float) -> float:
    """Multiply two numbers and return the result."""
    return a * b

print("Tool name:", multiply_numbers.name)
print("Tool description:", multiply_numbers.description)

# Test the tool directly
print("Test result:", multiply_numbers.invoke({"a": 12, "b": 8}))
````

````python
# Connect the calculator tool to Groq
llm_with_tools = llm.bind_tools([multiply_numbers])

# Ask a question that requires multiplication
response = llm_with_tools.invoke(
    "What is 125 multiplied by 24? Use the calculator tool."
)

print(response.tool_calls)
````

````python
# Execute the tool call and send the result back to Groq

from langchain_core.messages import ToolMessage

if response.tool_calls:
    tool_call = response.tool_calls[0]

    # Run the requested tool
    tool_result = multiply_numbers.invoke(tool_call["args"])

    print("Tool result:", tool_result)

    # Send the result back to the model
    tool_message = ToolMessage(
        content=str(tool_result),
        tool_call_id=tool_call["id"]
    )

    final_response = llm_with_tools.invoke([
        {"role": "user", "content": "What is 125 multiplied by 24?"},
        response,
        tool_message
    ])

    print("Final answer:", final_response.content)
else:
    print("The model did not request a tool call. Please run Step 13 again.")
````

````python
from langchain_core.messages import ToolMessage

tools_by_name = {
    multiply_numbers.name: multiply_numbers
}

def run_tool_agent(question):
    messages = [
        {"role": "user", "content": question}
    ]

    while True:
        ai_message = llm.bind_tools(
            [multiply_numbers]
        ).invoke(messages)

        messages.append(ai_message)

        # Stop if the model no longer requests a tool
        if not ai_message.tool_calls:
            return ai_message.content

        # Execute each requested tool
        for tool_call in ai_message.tool_calls:
            selected_tool = tools_by_name[tool_call["name"]]
            result = selected_tool.invoke(tool_call["args"])

            messages.append(
                ToolMessage(
                    content=str(result),
                    tool_call_id=tool_call["id"]
                )
            )

answer = run_tool_agent(
    "Calculate 125 multiplied by 24 using the calculator tool."
)

print("Agent answer:", answer)
````

````python
from langchain_core.tools import tool

@tool
def add_numbers(a: float, b: float) -> float:
    """Add two numbers and return the result."""
    return a + b

# Give the agent both tools
available_tools = [multiply_numbers, add_numbers]

tools_by_name = {
    tool.name: tool for tool in available_tools
}

print("Available tools:", list(tools_by_name.keys()))
print("Addition test:", add_numbers.invoke({"a": 15, "b": 25}))
print("Multiplication test:", multiply_numbers.invoke({"a": 12, "b": 8}))
````

````python
from langchain_core.messages import ToolMessage

def run_tool_agent(question):
    messages = [
        {"role": "user", "content": question}
    ]

    while True:
        ai_message = llm.bind_tools(available_tools).invoke(messages)
        messages.append(ai_message)

        if not ai_message.tool_calls:
            return ai_message.content

        for tool_call in ai_message.tool_calls:
            selected_tool = tools_by_name[tool_call["name"]]
            result = selected_tool.invoke(tool_call["args"])

            print(f"Tool used: {tool_call['name']}")
            print(f"Tool result: {result}")

            messages.append(
                ToolMessage(
                    content=str(result),
                    tool_call_id=tool_call["id"]
                )
            )

# Test the agent
answer = run_tool_agent(
    "First add 125 and 75, then multiply that result by 4."
)

print("\nFinal answer:", answer)
````

````python
from langchain_core.tools import tool
import wikipedia

@tool
def wikipedia_search(query: str) -> str:
    """Search Wikipedia for information about a person, place, event, or topic."""
    try:
        results = wikipedia.search(query, results=3)

        if not results:
            return "No Wikipedia results found."

        summaries = []

        for title in results[:2]:
            try:
                summary = wikipedia.summary(
                    title,
                    sentences=3,
                    auto_suggest=False
                )
                summaries.append(f"{title}: {summary}")
            except Exception:
                continue

        return "\n\n".join(summaries) or "No summaries could be retrieved."

    except Exception as e:
        return f"Wikipedia search error: {e}"

# Test the tool directly
print(wikipedia_search.invoke({"query": "LangChain"}))
````

````python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

# Start a fresh conversation
conversation_history = []

def chat_with_memory(user_input):
    messages = [
        SystemMessage(
            content="You are a friendly assistant. Remember details from the conversation history."
        )
    ]

    # Add previous messages
    messages.extend(conversation_history)

    # Add the new user message
    messages.append(HumanMessage(content=user_input))

    # Ask Groq
    response = llm.invoke(messages)

    # Save both sides of the conversation
    conversation_history.append(HumanMessage(content=user_input))
    conversation_history.append(AIMessage(content=response.content))

    return response.content

print(chat_with_memory("My favourite programming language is Python."))
print(chat_with_memory("What is my favourite programming language?"))
````

````python
from langchain_community.document_loaders import WebBaseLoader

urls = [
    "https://blog.gopenai.com/paper-review-llama-2-open-foundation-and-fine-tuned-chat-models-23e539522acb",
    "https://www.mosaicml.com/blog/mpt-7b",
    "https://stability.ai/blog/stability-ai-launches-the-first-of-its-stablelm-suite-of-language-models",
    "https://lmsys.org/blog/2023-03-30-vicuna/"
]

loader = WebBaseLoader(
    web_paths=urls,
    requests_kwargs={"timeout": 20}
)

docs = loader.load()

print("Number of pages loaded:", len(docs))

for i, doc in enumerate(docs):
    print(f"\nPage {i + 1}")
    print("Source:", doc.metadata.get("source"))
    print("Characters:", len(doc.page_content))
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

all_splits = text_splitter.split_documents(docs)

print("Total chunks created:", len(all_splits))

# Inspect the first chunk
print("\nFirst chunk preview:\n")
print(all_splits[0].page_content[:700])

print("\nSource URL:")
print(all_splits[0].metadata.get("source"))
````

````python
# Keep only pages with meaningful content
valid_docs = []

for doc in docs:
    text = doc.page_content.lower()

    if (
        len(doc.page_content) > 1500
        and "attention required! | cloudflare" not in text
        and "you have been blocked" not in text
    ):
        valid_docs.append(doc)

print("Original pages:", len(docs))
print("Usable pages:", len(valid_docs))

for doc in valid_docs:
    print(
        "\nSource:", doc.metadata.get("source"),
        "\nCharacters:", len(doc.page_content)
    )

# Recreate chunks using only usable pages
all_splits = text_splitter.split_documents(valid_docs)

print("\nTotal valid chunks:", len(all_splits))
````

````python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# Convert website text into numerical embeddings
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)

# Store the website chunks in FAISS
vectorstore = FAISS.from_documents(all_splits, embeddings)

print("Indexed chunks:", len(all_splits))
print("Vector database created successfully!")
````

````python
# Create a retriever that returns the 4 most relevant text chunks
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 4}
)

# Ask a test question
question = "What are the key features of MPT-7B?"

retrieved_docs = retriever.invoke(question)

print("Retrieved documents:", len(retrieved_docs))

for i, doc in enumerate(retrieved_docs, start=1):
    print(f"\n--- Document {i} ---")
    print("Source:", doc.metadata.get("source", "Unknown"))
    print(doc.page_content[:500])
````

````python
from operator import itemgetter

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Format retrieved website content
def format_docs(docs):
    return "\n\n".join(
        f"Source: {doc.metadata.get('source', 'Unknown')}\n"
        f"{doc.page_content}"
        for doc in docs
    )

# Create the prompt
rag_prompt = ChatPromptTemplate.from_template("""
You are a helpful website assistant.
Answer the question using only the provided context.
If the answer is not available in the context, say you don't know.

Context:
{context}

Question:
{question}

Answer:
""")

# Connect retrieval, prompt, Groq LLM, and output parser
rag_chain = (
    {
        "context": itemgetter("question") | retriever | format_docs,
        "question": itemgetter("question")
    }
    | rag_prompt
    | llm
    | StrOutputParser()
)

print("RAG chatbot chain created successfully!")
````

````python
question = "What are the key features of MPT-7B?"

answer = rag_chain.invoke({"question": question})

print("Question:", question)
print("\nAnswer:", answer)
````

````python
def answer_question_with_sources(question):
    # Retrieve relevant website content
    docs = retriever.invoke(question)

    # Prepare context
    context = format_docs(docs)

    # Generate answer
    answer = (rag_prompt | llm | StrOutputParser()).invoke({
        "context": context,
        "question": question
    })

    # Collect source URLs
    sources = list(dict.fromkeys(
        doc.metadata.get("source", "Unknown")
        for doc in docs
    ))

    return {
        "answer": answer,
        "sources": sources
    }

# Test the function
result = answer_question_with_sources(
    "What are the key features of MPT-7B?"
)

print("Answer:\n", result["answer"])
print("\nSources:")
for source in result["sources"]:
    print("-", source)
````

````python
print("Website RAG Chatbot is ready!")
print("Ask questions about the loaded websites.")
print("Type 'exit' to stop.\n")

while True:
    question = input("You: ").strip()

    if question.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    if not question:
        continue

    try:
        result = answer_question_with_sources(question)

        print("\nChatbot:", result["answer"])
        print("\nSources:")
        for source in result["sources"]:
            print("-", source)

        print("\n" + "-" * 60 + "\n")

    except Exception as e:
        print("An error occurred:", e)
````

````python
from langchain_core.messages import HumanMessage, AIMessage

chat_history = []

def chat_with_website(question):
    global chat_history

    docs = retriever.invoke(question)
    context = format_docs(docs)

    history_text = "\n".join(
        f"{'User' if isinstance(msg, HumanMessage) else 'Assistant'}: {msg.content}"
        for msg in chat_history[-6:]
    )

    prompt_with_history = ChatPromptTemplate.from_template("""
You are a helpful website assistant.
Use the website context and conversation history to answer.
If the context does not contain the answer, say you don't know.

Conversation history:
{history}

Website context:
{context}

Current question:
{question}

Answer:
""")

    answer = (
        prompt_with_history | llm | StrOutputParser()
    ).invoke({
        "history": history_text or "No previous conversation.",
        "context": context,
        "question": question
    })

    chat_history.extend([
        HumanMessage(content=question),
        AIMessage(content=answer)
    ])

    return {
        "answer": answer,
        "sources": list(dict.fromkeys(
            doc.metadata.get("source", "Unknown") for doc in docs
        ))
    }

print("Conversational website chatbot is ready!")
````

````python
# First question
result1 = chat_with_website(
    "What are the key features of MPT-7B?"
)

print("Question 1: What are the key features of MPT-7B?")
print("\nAnswer 1:", result1["answer"])

# Follow-up question
result2 = chat_with_website(
    "Can you explain its advantages in simple words?"
)

print("\nQuestion 2: Can you explain its advantages in simple words?")
print("\nAnswer 2:", result2["answer"])
````

````python
from langchain_community.document_loaders import WebBaseLoader

url = "https://lmsys.org/blog/2023-03-30-vicuna/"

loader = WebBaseLoader(url)
documents = loader.load()

print("Number of documents:", len(documents))
print("Document preview:")
print(documents[0].page_content[:1000])
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

chunks = text_splitter.split_documents(documents)

print("Total chunks:", len(chunks))
print("\nFirst chunk preview:\n")
print(chunks[0].page_content[:500])
````

````python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

sample_text = chunks[0].page_content
vector = embeddings.embed_query(sample_text)

print("Vector dimensions:", len(vector))
print("First 10 values:", vector[:10])
````

````python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)

print("FAISS vector database created successfully!")
print("Stored chunks:", len(chunks))
````

````python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

question = "What is the Vicuna language model?"

relevant_docs = retriever.invoke(question)

print("Retrieved documents:", len(relevant_docs))

for i, doc in enumerate(relevant_docs, start=1):
    print(f"\n--- Result {i} ---")
    print(doc.page_content[:500])
````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

rag_prompt = ChatPromptTemplate.from_template("""
Answer the question using only the context provided below.
If the answer is not in the context, say you don't know.

Context:
{context}

Question:
{question}
""")

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    }
    | rag_prompt
    | llm
    | StrOutputParser()
)

answer = rag_chain.invoke("What is the Vicuna language model?")
print(answer)
````

````python
def answer_question_with_sources(question):
    relevant_docs = retriever.invoke(question)

    context = format_docs(relevant_docs)

    prompt_value = rag_prompt.invoke({
        "context": context,
        "question": question
    })

    response = llm.invoke(prompt_value)

    print("ANSWER:\n")
    print(response.content)

    print("\nSOURCES:\n")
    for i, doc in enumerate(relevant_docs, start=1):
        print(f"{i}. {doc.metadata.get('source', 'Source unavailable')}")

answer_question_with_sources(
    "What are the main features of Vicuna?"
)
````

````python
import os, sys, textwrap, json, subprocess, time, re
from pathlib import Path

import torch
from IPython.display import display, Markdown

from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

print("Python:", sys.version.split()[0])
print("Torch:", torch.__version__)
print("GPU available:", torch.cuda.is_available())
````

````python
from google.colab import userdata

GROQ_API_KEY = userdata.get("GROQ_API_KEY")
if not GROQ_API_KEY:
    raise ValueError("Add GROQ_API_KEY in Colab Secrets and enable notebook access.")

os.environ["GROQ_API_KEY"] = GROQ_API_KEY
GROQ_MODEL = "openai/gpt-oss-20b"  # Change this if this model is unavailable in your Groq account.
os.environ["GROQ_MODEL"] = GROQ_MODEL

print("Groq API key loaded securely.")
print("Selected model:", GROQ_MODEL)
````

````python
URLS = [
    "https://blog.gopenai.com/paper-review-llama-2-open-foundation-and-fine-tuned-chat-models-23e539522acb",
    "https://www.mosaicml.com/blog/mpt-7b",
    "https://stability.ai/blog/stability-ai-launches-the-first-of-their-stablelm-suite-of-language-models",
    "https://lmsys.org/blog/2023-03-30-vicuna/",
]

print(f"URLs configured: {len(URLS)}")
for url in URLS:
    print("-", url)
````

````python
documents = []
failed_urls = []

for url in URLS:
    try:
        loader = WebBaseLoader(web_paths=(url,))
        loaded = loader.load()
        if loaded and loaded[0].page_content.strip():
            documents.extend(loaded)
            print(f"Loaded: {url} ({len(loaded[0].page_content):,} characters)")
        else:
            failed_urls.append(url)
            print(f"No readable text: {url}")
    except Exception as e:
        failed_urls.append(url)
        print(f"Failed: {url} -> {type(e).__name__}: {e}")

if not documents:
    raise RuntimeError("No website text was extracted. Check the URLs or try other pages.")

print("\nDocuments loaded:", len(documents))
print("Failed URLs:", len(failed_urls))
````

````python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    separators=["\n\n", "\n", ". ", " ", ""]
)

text_chunks = text_splitter.split_documents(documents)
print("Total chunks:", len(text_chunks))
print("First chunk preview:\n")
print(text_chunks[0].page_content[:1000])
````

````python
embedding_device = "cuda" if torch.cuda.is_available() else "cpu"
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": embedding_device},
    encode_kwargs={"normalize_embeddings": True},
)

sample_embedding = embeddings.embed_query("What is a large language model?")
print("Embedding dimension:", len(sample_embedding))
print("Embedding device:", embedding_device)
````

````python
vectorstore = FAISS.from_documents(text_chunks, embeddings)
retriever = vectorstore.as_retriever(search_kwargs={"k": 4})
print("FAISS knowledge base created.")
````

````python
llm = ChatGroq(
    model=GROQ_MODEL,
    temperature=0.2,
    max_tokens=1024,
    api_key=GROQ_API_KEY,
)

# Small connection test
test_response = llm.invoke("Reply with exactly: Groq connection successful.")
print(test_response.content)
````

````python
rag_prompt = ChatPromptTemplate.from_messages([
    ("system", """You are a helpful website research assistant.
Answer the user's question using ONLY the supplied context.
If the context does not contain the answer, clearly say that the available website content does not provide enough information.
Do not invent facts. Keep the answer clear and structured.

Context:
{context}
"""),
    ("human", "Question: {question}")
])


def answer_question(question: str, k: int = 4):
    question = question.strip()
    if not question:
        raise ValueError("Please enter a question.")

    retrieved_docs = vectorstore.similarity_search(question, k=k)
    if not retrieved_docs:
        return {"answer": "I could not find relevant information in the indexed pages.", "sources": []}

    context_parts = []
    sources = []
    for i, doc in enumerate(retrieved_docs, start=1):
        source = doc.metadata.get("source") or doc.metadata.get("url") or "Unknown source"
        context_parts.append(f"[Document {i}] Source: {source}\n{doc.page_content}")
        if source != "Unknown source" and source not in sources:
            sources.append(source)

    context = "\n\n".join(context_parts)
    messages = rag_prompt.format_messages(context=context, question=question)
    response = llm.invoke(messages)

    return {"answer": response.content, "sources": sources, "retrieved_documents": retrieved_docs}

print("RAG function ready.")
````

````python
example_questions = [
    "How good is Vicuna?",
    "How does Llama 2 compare with other models?",
    "What is StableLM?",
    "Can you share details about the MPT-7B model?",
]

for q in example_questions:
    result = answer_question(q)
    print("\nQUESTION:", q)
    print("ANSWER:", textwrap.fill(result["answer"], width=100))
    print("SOURCES:", result["sources"])
````

````python
while True:
    user_question = input("Ask about the indexed websites (or type 'exit'): ").strip()
    if user_question.lower() == "exit":
        print("Chat ended.")
        break
    if not user_question:
        continue

    result = answer_question(user_question)
    print("\nAnswer:\n", result["answer"])
    print("\nSources:")
    for source in result["sources"]:
        print("-", source)
    print("\n" + "=" * 80)
````

````python
%%writefile /content/app.py
import os
import streamlit as st
from langchain_groq import ChatGroq
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import ChatPromptTemplate

st.set_page_config(page_title="Website RAG Chatbot", page_icon="🌐", layout="wide")
st.title("🌐 Chat with Your Websites")
st.caption("Retrieval-Augmented Generation powered by Groq, Hugging Face and FAISS")

@st.cache_resource
def load_components():
    api_key = os.environ.get("GROQ_API_KEY")
    if not api_key:
        raise RuntimeError("GROQ_API_KEY is missing from the environment.")
    model_name = os.environ.get("GROQ_MODEL", "openai/gpt-oss-20b")
    embedder = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2",
        model_kwargs={"device": "cpu"},
        encode_kwargs={"normalize_embeddings": True},
    )
    db = FAISS.load_local(
        "/content/website_faiss_index",
        embedder,
        allow_dangerous_deserialization=True,
    )
    chat_model = ChatGroq(model=model_name, temperature=0.2, max_tokens=1024, api_key=api_key)
    return db, chat_model

try:
    db, chat_model = load_components()
except Exception as exc:
    st.error(f"Could not load chatbot: {exc}")
    st.stop()

prompt = ChatPromptTemplate.from_messages([
    ("system", """You are a helpful website research assistant. Answer using only the context below.
If the answer is not in the context, say that the indexed pages do not provide enough information.
Do not invent facts.

Context:
{context}
"""),
    ("human", "Question: {question}")
])

if "messages" not in st.session_state:
    st.session_state.messages = []

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.markdown(msg["content"])
        if msg.get("sources"):
            with st.expander("Sources"):
                for src in msg["sources"]:
                    st.write(src)

question = st.chat_input("Ask a question about the websites...")
if question:
    st.session_state.messages.append({"role": "user", "content": question})
    with st.chat_message("user"):
        st.markdown(question)

    with st.chat_message("assistant"):
        with st.spinner("Searching the website knowledge base..."):
            docs = db.similarity_search(question, k=4)
            context = "\n\n".join(
                f"Source: {d.metadata.get('source', 'Unknown')}\n{d.page_content}" for d in docs
            )
            sources = list(dict.fromkeys(
                d.metadata.get("source", "Unknown") for d in docs
            ))
            response = chat_model.invoke(prompt.format_messages(context=context, question=question))
            st.markdown(response.content)
            if sources:
                with st.expander("Sources"):
                    for src in sources:
                        st.write(src)
    st.session_state.messages.append({"role": "assistant", "content": response.content, "sources": sources})
````

# Falcon 7b chroma db retriever

````python
import torch
import transformers
import langchain
import chromadb
import sentence_transformers
import bitsandbytes
import langchain_huggingface
import langchain_chroma
import pypdf

print("All required libraries imported successfully!")

print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("LangChain:", langchain.__version__)
print("ChromaDB:", chromadb.__version__)
print("CUDA available:", torch.cuda.is_available())
````

````python
!wget https://github.com/Shafi2016/Youtube/raw/main/stock_market_june_2023.zip -O stock_market_june_2023.zip

!unzip -o stock_market_june_2023.zip
````

````python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

# Load all text files from the folder
loader = DirectoryLoader(
    "./stock_market_june_2023/",
    glob="*.txt",
    loader_cls=TextLoader
)

documents = loader.load()

print("Documents loaded successfully!")
print("Total documents:", len(documents))
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

texts = text_splitter.split_documents(documents)

print("Document splitting completed!")
print("Total chunks:", len(texts))
````

````python
print("Chunk content:")
print(texts[2].page_content)

print("\nChunk metadata:")
print(texts[2].metadata)
````

````python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="intfloat/e5-large-v2",
    model_kwargs={"device": "cuda"},
    encode_kwargs={"normalize_embeddings": True}
)

print("Embedding model loaded successfully!")
````

````python
import time

test_questions = [
    "Which companies are highlighted in the dataset for potential stock-market growth?",
    "How much money did Microsoft invest in OpenAI?",
    "Why did Shopify announce layoffs in 2023?"
]

mistral_results = []

for i, question in enumerate(test_questions, 1):
    start_time = time.perf_counter()

    result = qa_chain.invoke({"query": question})

    end_time = time.perf_counter()
    response_time = end_time - start_time

    mistral_results.append({
        "Question": question,
        "Answer": result["result"],
        "Response Time (seconds)": round(response_time, 2),
        "Sources": list(set(
            doc.metadata.get("source", "Unknown")
            for doc in result["source_documents"]
        ))
    })

    print(f"\nQuestion {i}: {question}")
    print(f"Response time: {response_time:.2f} seconds")
    print(f"Sources: {mistral_results[-1]['Sources']}")
    print("-" * 70)

print("\nMistral evaluation completed.")
````

````python
from langchain_classic.chains import RetrievalQA

# Create retriever
repro_retriever = vectorstore_check.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

# Create RAG chain
repro_qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=repro_retriever,
    return_source_documents=True
)

# Test question
question = "How much money did Microsoft invest in OpenAI?"

result = repro_qa_chain.invoke({"query": question})

print("QUESTION:", question)
print("\nANSWER:")
print(result["result"])

print("\nRETRIEVED SOURCES:")
for doc in result["source_documents"]:
    print("-", doc.metadata.get("source"))

print("\nRAG PIPELINE TEST COMPLETED!")
````

````python
%%writefile app.py

import os
import streamlit as st

from langchain_groq import ChatGroq
from langchain_community.document_loaders import UnstructuredURLLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain.chains import create_retrieval_chain
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate


st.set_page_config(
    page_title="Victoria On Move AI Assistant",
    page_icon="🚚"
)

st.title("🚚 Victoria On Move AI Assistant")
st.caption("Ask questions about moving services, prices and bookings.")


@st.cache_resource
def create_rag_chain():

    urls = [
        "https://www.victoriaonmove.com.au/local-removalists.html",
        "https://victoriaonmove.com.au/index.html",
        "https://victoriaonmove.com.au/contact.html"
    ]

    loader = UnstructuredURLLoader(urls=urls)
    data = loader.load()

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200
    )

    chunks = splitter.split_documents(data)

    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

    vectorstore = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings,
        collection_name="victoria_on_move_app"
    )

    retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k": 6}
    )

    llm = ChatGroq(
        model="openai/gpt-oss-120b",
        temperature=0.4,
        max_tokens=500
    )

    system_prompt = (
        "You are a helpful customer support assistant "
        "for Victoria On Move. "
        "Answer using only the retrieved context. "
        "If the answer is unavailable, say you don't know. "
        "Never invent prices or company information. "
        "Keep answers concise and within three sentences.\n\n"
        "Context:\n{context}"
    )

    prompt = ChatPromptTemplate.from_messages([
        ("system", system_prompt),
        ("human", "{input}")
    ])

    qa_chain = create_stuff_documents_chain(llm, prompt)

    return create_retrieval_chain(retriever, qa_chain)


if not os.environ.get("GROQ_API_KEY"):
    st.error("GROQ_API_KEY is not configured.")
    st.stop()

try:
    rag_chain = create_rag_chain()
    st.success("AI assistant is ready!")

except Exception as e:
    st.error(f"Error initializing RAG application: {e}")
    st.stop()


if "messages" not in st.session_state:
    st.session_state.messages = []


for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])


question = st.chat_input("Ask about moving services, prices, etc.")

if question:

    st.session_state.messages.append({
        "role": "user",
        "content": question
    })

    with st.chat_message("user"):
        st.markdown(question)

    with st.chat_message("assistant"):

        with st.spinner("Searching website information..."):

            try:
                response = rag_chain.invoke({
                    "input": question
                })

                answer = response["answer"]

                st.markdown(answer)

                sources = list(dict.fromkeys(
                    doc.metadata.get("source", "")
                    for doc in response["context"]
                    if doc.metadata.get("source")
                ))

                with st.expander("View sources"):
                    for source in sources:
                        st.write(source)

                st.session_state.messages.append({
                    "role": "assistant",
                    "content": answer
                })

            except Exception as e:
                st.error(f"Error generating answer: {e}")
````

````python
import os

project_dir = "/content/rainfall_streamlit_project"

requirements = """streamlit
langchain==1.4.8
langchain-classic==1.0.8
langchain-community==0.4.2
langchain-core==1.7.4
langchain-text-splitters==1.1.3
langchain-huggingface==1.2.3
langchain-chroma==1.1.0
langchain-groq==1.1.3
chromadb==1.5.9
sentence-transformers==5.7.0
"""

app_code = r'''
import os
import streamlit as st

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain

st.set_page_config(
    page_title="Rainfall Research Assistant",
    page_icon="🌧️",
    layout="wide"
)

st.title("🌧️ Rainfall Research Assistant")
st.caption("Ask questions about the research paper: Development of Multiple Combined Regression Methods for Rainfall Measurement.")

@st.cache_resource
def load_rag():
    db_path = os.path.join(os.path.dirname(__file__), "rainfall_chroma_db")

    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

    vectorstore = Chroma(
        persist_directory=db_path,
        collection_name="rainfall_research_v2",
        embedding_function=embeddings
    )

    retriever = vectorstore.as_retriever(
        search_kwargs={"k": 12}
    )

    llm = ChatGroq(
        model="openai/gpt-oss-120b",
        temperature=0.2,
        max_tokens=1200,
        groq_api_key=st.secrets["GROQ_API_KEY"]
    )

    document_prompt = PromptTemplate(
        input_variables=["page_content", "page"],
        template="PDF PAGE: {page}\nDOCUMENT CONTENT:\n{page_content}"
    )

    prompt = ChatPromptTemplate.from_messages([
        (
            "system",
            """You are a precise research assistant answering questions
            using the supplied rainfall research paper context.

            Follow these rules:
            1. Read all retrieved context before answering.
            2. Search across all supplied chunks for the answer.
            3. Combine information from continuation chunks when needed.
            4. For tables, match each value to the correct model and metric.
            5. Preserve exact numerical values from the paper.
            6. Include the PDF page number for factual answers.
            7. If the paper does not provide the answer, clearly say so.
            8. Never invent facts, values, hardware details, or experiments.
            9. If the paper contains conflicting statements, explain the
               discrepancy instead of silently choosing one.
            10. For model comparisons, consider every relevant row provided.

            Research context:
            {context}"""
        ),
        ("human", "{input}")
    ])

    document_chain = create_stuff_documents_chain(
        llm,
        prompt,
        document_prompt=document_prompt
    )

    return create_retrieval_chain(retriever, document_chain)

try:
    rag_chain = load_rag()
    st.success("Research database loaded successfully.")
except Exception as e:
    st.error(f"Could not load the chatbot: {e}")
    st.stop()

if "messages" not in st.session_state:
    st.session_state.messages = []

with st.sidebar:
    st.subheader("About this chatbot")
    st.write("Answers are generated from the rainfall research paper and its indexed document database.")
    if st.button("Clear conversation"):
        st.session_state.messages = []
        st.rerun()

for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

question = st.chat_input("Ask a question about the rainfall research paper...")

if question:
    st.session_state.messages.append({
        "role": "user",
        "content": question
    })

    with st.chat_message("user"):
        st.markdown(question)

    with st.chat_message("assistant"):
        with st.spinner("Searching the research paper..."):
            try:
                response = rag_chain.invoke({"input": question})
                answer = response["answer"]
                st.markdown(answer)

                source_docs = response.get("context", [])
                if source_docs:
                    with st.expander("View retrieved sources"):
                        seen = set()
                        for doc in source_docs:
                            page = doc.metadata.get("page")
                            source = doc.metadata.get("source", "Research paper")
                            key = (source, page, doc.page_content[:100])

                            if key not in seen:
                                seen.add(key)
                                st.markdown(
                                    f"**Source:** {source} | "
                                    f"**PDF page:** {page}"
                                )
                                st.caption(doc.page_content[:500])

                st.session_state.messages.append({
                    "role": "assistant",
                    "content": answer
                })

            except Exception as e:
                st.error(f"An error occurred: {e}")
'''

with open(os.path.join(project_dir, "requirements.txt"), "w") as f:
    f.write(requirements)

with open(os.path.join(project_dir, "app.py"), "w") as f:
    f.write(app_code)

print("Created requirements.txt")
print("Created app.py")
print("\nProject files:")
for item in sorted(os.listdir(project_dir)):
    print("-", item)
````

````python
from trl import SFTConfig, SFTTrainer

sft_config = SFTConfig(
    output_dir="./results",
    num_train_epochs=1,
    per_device_train_batch_size=4,
    gradient_accumulation_steps=1,
    learning_rate=2e-4,
    optim="paged_adamw_32bit",
    fp16=True,
    logging_steps=25,
    save_steps=25,
    warmup_steps=10,
    weight_decay=0.001,
    max_grad_norm=0.3,
    lr_scheduler_type="constant",
    report_to="none",
    dataset_text_field="text",
    max_length=512,
    packing=False
)

trainer = SFTTrainer(
    model=model,
    args=sft_config,
    train_dataset=dataset,
    processing_class=tokenizer
)

print("SFT Trainer created successfully!")
print("Training examples:", len(dataset))
print("Maximum sequence length:", sft_config.max_length)
print("Packing:", sft_config.packing)
````

````python
# Convert only trainable LoRA parameters to FP32
for param in model.parameters():
    if param.requires_grad:
        param.data = param.data.to(torch.float32)

# Confirm the conversion
for name, param in model.named_parameters():
    if param.requires_grad:
        print("Trainable parameter:", name)
        print("New dtype:", param.dtype)
        break

# Rebuild the trainer with the existing FP16 configuration
trainer = SFTTrainer(
    model=model,
    args=sft_config,
    train_dataset=dataset,
    processing_class=tokenizer
)

print("\nTrainer rebuilt successfully!")
print("FP16:", sft_config.fp16)
print("BF16:", sft_config.bf16)
````

````python
from collections import Counter

dtype_counts = Counter()
bf16_parameters = []

for name, param in trainer.model.named_parameters():
    if param.requires_grad:
        dtype_counts[str(param.dtype)] += 1

        if param.dtype == torch.bfloat16:
            bf16_parameters.append(name)

print("Trainable parameter dtype counts:")
print(dtype_counts)

print("\nNumber of BF16 trainable parameters:", len(bf16_parameters))

if bf16_parameters:
    print("\nFirst BF16 parameters:")
    for name in bf16_parameters[:10]:
        print(name)
````

````python
# Convert every trainable LoRA parameter to FP32
for name, param in trainer.model.named_parameters():
    if param.requires_grad:
        param.data = param.data.to(torch.float32)

# Verify all trainable parameter dtypes
from collections import Counter

dtype_counts = Counter(
    str(param.dtype)
    for param in trainer.model.parameters()
    if param.requires_grad
)

print("Trainable parameter dtype counts:")
print(dtype_counts)

assert all(
    param.dtype == torch.float32
    for param in trainer.model.parameters()
    if param.requires_grad
), "Some trainable parameters are still not FP32!"

print("\nSuccess: All trainable parameters are FP32.")
````

````python
from trl import SFTConfig, SFTTrainer

# Create training configuration without FP16/BF16 mixed-precision training
sft_config = SFTConfig(
    output_dir="./results",
    num_train_epochs=1,
    per_device_train_batch_size=4,
    gradient_accumulation_steps=1,
    learning_rate=2e-4,
    optim="paged_adamw_32bit",
    fp16=False,
    bf16=False,
    logging_steps=25,
    save_steps=25,
    warmup_steps=10,
    weight_decay=0.001,
    max_grad_norm=0.3,
    lr_scheduler_type="constant",
    report_to="none",
    dataset_text_field="text",
    max_length=512,
    packing=False
)

# Recreate trainer
trainer = SFTTrainer(
    model=model,
    args=sft_config,
    train_dataset=dataset,
    processing_class=tokenizer
)

# Convert trainable LoRA parameters to FP32 AFTER trainer creation
for param in trainer.model.parameters():
    if param.requires_grad:
        param.data = param.data.to(torch.float32)

# Verify configuration and parameter types
from collections import Counter

dtype_counts = Counter(
    str(param.dtype)
    for param in trainer.model.parameters()
    if param.requires_grad
)

print("FP16 enabled:", trainer.args.fp16)
print("BF16 enabled:", trainer.args.bf16)
print("Trainable parameter dtypes:", dtype_counts)
print("Trainer recreated successfully!")
````

````python
training_result = trainer.train()

print("\nFine-tuning completed!")
print("Training loss:", training_result.training_loss)
````

````python
# Save the fine-tuned LoRA adapter
adapter_path = "./Llama-2-7b-chat-finetune"

trainer.model.save_pretrained(adapter_path)
tokenizer.save_pretrained(adapter_path)

print("Fine-tuned LoRA adapter saved successfully!")
print("Location:", adapter_path)
````

````python
import torch

evaluation_results = []

for i, question in enumerate(evaluation_questions, 1):

    prompt = f"<s>[INST] {question} [/INST]"

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():

        # Original model
        original_output = base_model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

        # Fine-tuned model
        fine_tuned_output = fine_tuned_model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    original_answer = tokenizer.decode(
        original_output[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    fine_tuned_answer = tokenizer.decode(
        fine_tuned_output[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    evaluation_results.append({
        "Question": question,
        "Original Answer": original_answer,
        "Fine-tuned Answer": fine_tuned_answer
    })

    print(f"\n{'='*70}")
    print(f"QUESTION {i}: {question}")
    print("\nORIGINAL LLAMA 2:")
    print(original_answer)
    print("\nFINE-TUNED LLAMA 2:")
    print(fine_tuned_answer)

print("\n\nEvaluation generation completed!")
````

````python
# Calculate stock performance metrics

latest_price = stock_data["Close"].iloc[-1]
previous_price = stock_data["Close"].iloc[-2]
year_start_price = stock_data["Close"].iloc[0]

daily_change = ((latest_price - previous_price) / previous_price) * 100

yearly_return = (
    (latest_price - year_start_price) / year_start_price
) * 100

highest_price = stock_data["High"].max()
lowest_price = stock_data["Low"].min()

average_volume = stock_data["Volume"].mean()

print("NVIDIA STOCK PERFORMANCE")
print("-" * 35)
print(f"Latest closing price: ${latest_price:.2f}")
print(f"Previous closing price: ${previous_price:.2f}")
print(f"Daily change: {daily_change:.2f}%")
print(f"1-year return: {yearly_return:.2f}%")
print(f"52-week high: ${highest_price:.2f}")
print(f"52-week low: ${lowest_price:.2f}")
print(f"Average daily volume: {average_volume:,.0f}")
````

````python
# Calculate Simple Moving Averages
stock_data["SMA_20"] = stock_data["Close"].rolling(window=20).mean()
stock_data["SMA_50"] = stock_data["Close"].rolling(window=50).mean()

# Calculate RSI (14-day)
delta = stock_data["Close"].diff()

gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

avg_gain = gain.rolling(window=14).mean()
avg_loss = loss.rolling(window=14).mean()

rs = avg_gain / avg_loss
stock_data["RSI_14"] = 100 - (100 / (1 + rs))

# Display latest indicators
print("NVIDIA TECHNICAL INDICATORS")
print("-" * 35)
print(f"Latest Close: ${stock_data['Close'].iloc[-1]:.2f}")
print(f"SMA 20: ${stock_data['SMA_20'].iloc[-1]:.2f}")
print(f"SMA 50: ${stock_data['SMA_50'].iloc[-1]:.2f}")
print(f"RSI 14: {stock_data['RSI_14'].iloc[-1]:.2f}")
````

````python
# Inspect raw financial data and data types

print("FINANCIAL DATA VALIDATION")
print("=" * 50)

fields_to_check = [
    "revenueGrowth",
    "grossMargins",
    "operatingMargins",
    "profitMargins",
    "debtToEquity",
    "returnOnEquity",
    "trailingPE",
    "forwardPE"
]

for field in fields_to_check:
    value = fundamentals.get(field)

    print(
        f"{field}: {value} "
        f"| Python type: {type(value).__name__}"
    )
````

````python
validation_rules = {
    "revenueGrowth": (-1, 10),
    "grossMargins": (0, 1),
    "operatingMargins": (0, 1),
    "profitMargins": (0, 1),
    "returnOnEquity": (-5, 5),
    "trailingPE": (0, 200),
    "forwardPE": (0, 200),
    "debtToEquity": (0, 1000)
}

print("AUTOMATED VALIDATION RESULTS")
print("=" * 55)

for field, (minimum, maximum) in validation_rules.items():
    value = fundamentals.get(field)

    if value is None:
        status = "MISSING"
    elif not isinstance(value, (int, float)):
        status = "INVALID TYPE"
    elif not minimum <= value <= maximum:
        status = "REVIEW REQUIRED"
    else:
        status = "PASSED"

    print(f"{field:20} | {value!s:12} | {status}")
````

````python
# Retrieve competitor stock data

competitors = {
    "NVIDIA": "NVDA",
    "AMD": "AMD",
    "Intel": "INTC"
}

competitor_data = {}

for company_name, ticker_symbol in competitors.items():

    ticker = yf.Ticker(ticker_symbol)

    history = ticker.history(period="1y")
    info = ticker.info

    if history.empty:
        print(f"Data unavailable for {company_name}")
        continue

    latest = history["Close"].iloc[-1]
    first = history["Close"].iloc[0]

    competitor_data[company_name] = {
        "Ticker": ticker_symbol,
        "Latest Price": round(float(latest), 2),
        "One-Year Return (%)": round(
            float((latest - first) / first * 100), 2
        ),
        "Market Cap": info.get("marketCap"),
        "Revenue": info.get("totalRevenue"),
        "Revenue Growth (%)": (
            round(info["revenueGrowth"] * 100, 2)
            if info.get("revenueGrowth") is not None else None
        ),
        "Gross Margin (%)": (
            round(info["grossMargins"] * 100, 2)
            if info.get("grossMargins") is not None else None
        ),
        "Trailing P/E": info.get("trailingPE")
    }

print("COMPETITOR DATA COLLECTION")
print("=" * 45)

display(pd.DataFrame(competitor_data).T)
````

````python

from GoogleNews import GoogleNews
import pandas as pd

companies = {
    "NVIDIA": "NVIDIA stock",
    "AMD": "AMD stock",
    "Intel": "Intel stock"
}

news_data = {}

for company, search_query in companies.items():

    google_news = GoogleNews(
        lang="en",
        region="US",
        period="7d"
    )

    google_news.search(search_query)

    results = google_news.result()

    news_data[company] = results

    print(f"\n{'='*50}")
    print(f"LATEST NEWS: {company}")
    print("="*50)

    if not results:
        print("No news found.")
    else:
        for article in results[:5]:
            print("Headline:", article.get("title"))
            print("Source:", article.get("media"))
            print("Date:", article.get("date"))
            print("Link:", article.get("link"))
            print("-" * 40)
````

````python

import feedparser
import pandas as pd
from urllib.parse import quote

companies = {
    "NVIDIA": "NVDA stock",
    "AMD": "AMD stock",
    "Intel": "INTC stock"
}

news_records = []

for company, query in companies.items():

    rss_url = (
        "https://feeds.finance.yahoo.com/rss/2.0/headline"
        f"?s={company == 'NVIDIA' and 'NVDA' or company == 'AMD' and 'AMD' or 'INTC'}"
        "&region=US&lang=en-US"
    )

    feed = feedparser.parse(rss_url)

    print(f"\n{'='*50}")
    print(f"{company} NEWS")
    print("="*50)

    if not feed.entries:
        print("No news available from this feed.")
        continue

    for article in feed.entries[:5]:

        record = {
            "Company": company,
            "Headline": article.get("title", "No title"),
            "Published": article.get("published", "Unknown"),
            "Link": article.get("link", "")
        }

        news_records.append(record)

        print("Headline:", record["Headline"])
        print("Published:", record["Published"])
        print("Link:", record["Link"])
        print("-" * 40)

news_df = pd.DataFrame(news_records)

print("\nTotal news articles collected:", len(news_df))
````

````python

import re
import pandas as pd

company_keywords = {
    "NVIDIA": [r"\bNVIDIA\b", r"\bNVDA\b"],
    "AMD": [r"\bAMD\b", r"\bAdvanced Micro Devices\b"],
    "Intel": [r"\bIntel\b", r"\bINTC\b"]
}

def check_relevance(row):
    headline = str(row["Headline"])

    keywords = company_keywords[row["Company"]]

    for keyword in keywords:
        if re.search(keyword, headline, re.IGNORECASE):
            return "Directly Relevant"

    return "Needs Review"

news_df["Relevance"] = news_df.apply(
    check_relevance,
    axis=1
)

print("NEWS RELEVANCE REPORT")
print("=" * 60)

display(
    news_df[
        ["Company", "Headline", "Published", "Relevance"]
    ]
)

print("\nRELEVANCE SUMMARY")
display(
    news_df.groupby(["Company", "Relevance"]).size()
    .unstack(fill_value=0)
)
````

````python

import re
import pandas as pd

company_names = {
    "NVIDIA": ["NVIDIA", "NVDA"],
    "AMD": ["AMD", "ADVANCED MICRO DEVICES"],
    "Intel": ["INTEL", "INTC"]
}

market_context_patterns = [
    r"\bfutures\b",
    r"\bstocks mixed\b",
    r"\band more\b",
    r"\bin focus\b"
]

def classify_news(row):
    headline = str(row["Headline"]).upper()
    company = row["Company"]

    target_found = any(
        re.search(r"\b" + re.escape(word) + r"\b", headline)
        for word in company_names[company]
    )

    if not target_found:
        return "Needs Review"

    if any(
        re.search(pattern, headline, re.IGNORECASE)
        for pattern in market_context_patterns
    ):
        return "Market Context"

    return "Directly Relevant"

news_df["Relevance"] = news_df.apply(
    classify_news, axis=1
)

print("IMPROVED NEWS CLASSIFICATION")
display(
    news_df[
        ["Company", "Headline", "Relevance"]
    ]
)

print("\nCLASSIFICATION SUMMARY")
display(
    news_df.groupby(["Company", "Relevance"])
    .size()
    .unstack(fill_value=0)
)

# Keep direct company news for the next AI analysis step
relevant_news_df = news_df[
    news_df["Relevance"] == "Directly Relevant"
].copy()

print(
    "\nDirect company news articles:",
    len(relevant_news_df)
)
````

````python

import gradio as gr
import pandas as pd

def get_company_summary(company):
    details = competitor_data[company]

    return f"""
# {company} Financial Overview

- **Ticker:** {details['Ticker']}
- **Latest Price:** ${details['Latest Price']}
- **One-Year Price Return:** {details['One-Year Return (%)']}%
- **Revenue:** ${details['Revenue'] / 1_000_000_000:.2f} Billion
- **Revenue Growth:** {details['Revenue Growth (%)']}%
- **Gross Margin:** {details['Gross Margin (%)']}%
- **Trailing P/E:** {details['Trailing P/E']}

*Source: Yahoo Finance via yfinance*
"""

with gr.Blocks(title="AI Financial Stock Analyzer") as dashboard:

    gr.Markdown(
        """
        # 📊 AI Financial Stock Analyzer
        ### Financial insights powered by Groq AI
        """
    )

    with gr.Row():
        company_dropdown = gr.Dropdown(
            choices=list(competitor_data.keys()),
            value="NVIDIA",
            label="Select Company"
        )

    company_output = gr.Markdown()

    company_dropdown.change(
        fn=get_company_summary,
        inputs=company_dropdown,
        outputs=company_output
    )

    gr.Markdown("## AI Financial Analysis")

    ai_output = gr.Markdown(
        "Select a company and generate its AI analysis."
    )

    def generate_company_analysis(company):
        prompt = f"""
        Analyze this financial data:

        {json.dumps(competitor_data[company], indent=2)}

        Explain:
        1. Stock performance
        2. Revenue growth
        3. Gross margin
        4. P/E valuation
        5. Missing information and risks

        Use only supplied data.
        Do not invent facts or give buy/sell recommendations.
        """

        result = llm.complete(prompt)
        return result.text

    analyze_button = gr.Button(
        "Generate AI Analysis",
        variant="primary"
    )

    analyze_button.click(
        fn=generate_company_analysis,
        inputs=company_dropdown,
        outputs=ai_output
    )

    dashboard.load(
        fn=get_company_summary,
        inputs=company_dropdown,
        outputs=company_output
    )

dashboard.launch(share=True)
````

````python

import json
import pandas as pd

def analyze_news_sentiment(company):

    articles = news_for_ai.get(company, [])

    if not articles:
        return pd.DataFrame(columns=[
            "Company", "Headline", "Sentiment"
        ])

    prompt = f"""
You are a cautious financial news sentiment classifier.

Classify each headline for {company}.

HEADLINES:
{json.dumps(articles, indent=2)}

For every headline, assign exactly one label:
Positive
Negative
Mixed
Insufficient Evidence

Rules:
- Use only the headline.
- Do not invent article details.
- Do not predict stock prices.
- Ambiguous headlines must be Insufficient Evidence.
- Return valid JSON only.
- Preserve the original headline.

Required JSON format:
{{
  "results": [
    {{
      "headline": "Original headline",
      "sentiment": "Mixed"
    }}
  ]
}}
"""

    response = llm.complete(prompt)

    try:
        result_text = response.text.strip()

        result_text = result_text.replace(
            "```json", ""
        ).replace("```", "").strip()

        parsed = json.loads(result_text)

        records = []

        for item in parsed["results"]:
            sentiment = item["sentiment"]

            if sentiment not in [
                "Positive",
                "Negative",
                "Mixed",
                "Insufficient Evidence"
            ]:
                sentiment = "Insufficient Evidence"

            records.append({
                "Company": company,
                "Headline": item["headline"],
                "Sentiment": sentiment
            })

        return pd.DataFrame(records)

    except Exception as error:
        print("Sentiment parsing error:", error)

        return pd.DataFrame(columns=[
            "Company", "Headline", "Sentiment"
        ])

print("Sentiment analysis function created successfully!")
print("Ready to analyse NVIDIA, AMD and Intel news.")
````

````python

sentiment_results = []

for company in ["NVIDIA", "AMD", "Intel"]:

    print(f"\nAnalysing {company} news...")

    company_sentiment = analyze_news_sentiment(company)

    if not company_sentiment.empty:
        sentiment_results.append(company_sentiment)

if sentiment_results:

    sentiment_df = pd.concat(
        sentiment_results,
        ignore_index=True
    )

    print("\nNEWS SENTIMENT RESULTS")
    display(sentiment_df)

    print("\nSENTIMENT SUMMARY")
    sentiment_summary = pd.crosstab(
        sentiment_df["Company"],
        sentiment_df["Sentiment"]
    )

    display(sentiment_summary)

else:
    print("No sentiment results were generated.")
````

````python

import matplotlib.pyplot as plt

sentiment_order = [
    "Positive",
    "Negative",
    "Mixed",
    "Insufficient Evidence"
]

sentiment_summary = pd.crosstab(
    sentiment_df["Company"],
    sentiment_df["Sentiment"]
).reindex(
    columns=sentiment_order,
    fill_value=0
)

ax = sentiment_summary.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6)
)

ax.set_title("AI Financial News Sentiment Comparison")
ax.set_xlabel("Company")
ax.set_ylabel("Number of Headlines")
ax.legend(title="News Sentiment")
ax.tick_params(axis="x", rotation=0)
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()

print("Sentiment chart generated successfully!")
````

````python

import json
from llama_index.core import Settings, Document, VectorStoreIndex
from llama_index.llms.groq import Groq
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

# 1. Connect Groq as the LlamaIndex language model
Settings.llm = Groq(
    model="openai/gpt-oss-120b",
    api_key=GROQ_API_KEY,
    temperature=0.2
)

# 2. Configure local sentence-transformer embeddings
Settings.embed_model = HuggingFaceEmbedding(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# 3. Convert competitor financial data into searchable documents
financial_documents = []

for company_name, company_data in competitor_data.items():
    financial_documents.append(
        Document(
            text=f"""
            Company: {company_name}

            Financial Data:
            {json.dumps(company_data, indent=2, default=str)}
            """,
            metadata={
                "company": company_name,
                "document_type": "competitor_financial_data"
            }
        )
    )

# 4. Add NVIDIA's detailed technical analysis
financial_documents.append(
    Document(
        text=f"""
        NVIDIA Technical Analysis:

        {json.dumps(stock_summary, indent=2, default=str)}
        """,
        metadata={
            "company": "NVIDIA",
            "document_type": "technical_analysis"
        }
    )
)

# 5. Add NVIDIA's validated fundamental data
financial_documents.append(
    Document(
        text=f"""
        NVIDIA Validated Fundamentals:

        {json.dumps(validated_fundamentals, indent=2, default=str)}
        """,
        metadata={
            "company": "NVIDIA",
            "document_type": "fundamentals"
        }
    )
)

# 6. Build the searchable vector index
financial_index = VectorStoreIndex.from_documents(financial_documents)

print("LlamaIndex integration completed successfully!")
print("Language model: Groq - openai/gpt-oss-120b")
print("Embedding model: all-MiniLM-L6-v2")
print("Financial documents indexed:", len(financial_documents))
print("Companies:", list(competitor_data.keys()))
````

````python

# Create the RAG query engine
query_engine = financial_index.as_query_engine(
    similarity_top_k=3,
    response_mode="compact"
)

# Test the retrieval system
question = """
Compare NVIDIA, AMD and Intel based on their revenue,
revenue growth and gross margins.
Use only the indexed financial data.
"""

response = query_engine.query(question)

print("USER QUESTION:")
print(question)

print("\nRAG RESPONSE:")
print(response)

print("\nRETRIEVED SOURCE DOCUMENTS:")

for i, source in enumerate(response.source_nodes, start=1):
    print(f"\nSource {i}")
    print("Company:", source.node.metadata.get("company"))
    print("Document type:", source.node.metadata.get("document_type"))
    print("Similarity score:", source.score)
````

````python
from llama_index.core import StorageContext, load_index_from_storage

# Load the saved financial index
storage_context = StorageContext.from_defaults(
    persist_dir="./financial_stock_index"
)

loaded_financial_index = load_index_from_storage(storage_context)

# Create a query engine
loaded_query_engine = loaded_financial_index.as_query_engine(
    similarity_top_k=5,
    response_mode="compact"
)

# Test the reloaded index
question = """
What financial information is available for NVIDIA, AMD and Intel?
Summarize their revenue, revenue growth and gross margins.
Use only the indexed documents. Do not invent missing values.
"""

response = loaded_query_engine.query(question)

print("RELOADED INDEX TEST")
print("=" * 60)
print(response)

print("\nRETRIEVED SOURCE DOCUMENTS")
print("=" * 60)

for i, source in enumerate(response.source_nodes, start=1):
    print(
        f"{i}. Company: {source.node.metadata.get('company')} | "
        f"Type: {source.node.metadata.get('document_type')} | "
        f"Score: {source.score:.3f}"
    )
````

````python
def ask_financial_rag(question):
    response = loaded_query_engine.query(question)

    print("\nFINANCIAL AI ANALYSIS")
    print("=" * 60)
    print(response)

    print("\nSOURCE DOCUMENTS")
    print("=" * 60)

    for i, source in enumerate(response.source_nodes, start=1):
        metadata = source.node.metadata

        print(f"\nSource {i}")
        print("Company:", metadata.get("company"))
        print("Document Type:", metadata.get("document_type"))
        print("Relevance Score:", round(source.score, 3))

    return response


# Test the reusable RAG assistant
ask_financial_rag(
    "What are NVIDIA's financial strengths and weaknesses "
    "based on the indexed data?"
)
````

````python
def ask_financial_rag(question):

    safe_question = f"""
    You are a cautious financial research assistant.

    Answer the following question using ONLY the retrieved indexed documents.

    FINANCIAL ANALYSIS RULES:

    1. Never invent financial figures or missing information.
    2. Distinguish reported data from your interpretation.
    3. Do not assume a company is undervalued based only on its P/E ratio.
    4. Do not assume high ROE automatically means efficient leverage.
    5. Do not interpret debt-to-equity unless its units and definition
       are confirmed in the retrieved documents.
    6. Do not claim a company can comfortably service its debt without
       sufficient cash-flow and debt information.
    7. Treat RSI as a technical indicator, not a guaranteed prediction.
    8. Clearly mention when the reporting period is not specified.
    9. If evidence is insufficient, explicitly say so.
    10. Never present investment advice or guaranteed predictions.

    Structure your answer using:
    - Financial Strengths
    - Financial Weaknesses and Risks
    - Information Not Available
    - Overall Assessment

    User Question:
    {question}
    """

    response = loaded_query_engine.query(safe_question)

    print("\nFINANCIAL AI ANALYSIS")
    print("=" * 60)
    print(response)

    print("\nSOURCE DOCUMENTS")
    print("=" * 60)

    for i, source in enumerate(response.source_nodes, start=1):
        metadata = source.node.metadata

        print(f"\nSource {i}")
        print("Company:", metadata.get("company"))
        print("Document Type:", metadata.get("document_type"))
        print("Relevance Score:", round(source.score, 3))

    return response


ask_financial_rag(
    "What are NVIDIA's financial strengths and weaknesses "
    "based on the indexed data?"
)
````

````python

import gradio as gr
import pandas as pd
import re
import traceback

# ==================================================
# 1. PREPARE COMPARISON DATA
# ==================================================

comparison_df = pd.DataFrame.from_dict(
    competitor_data,
    orient="index"
).reset_index()

comparison_df = comparison_df.rename(
    columns={"index": "Company"}
)

# ==================================================
# 2. COMBINED COMPANY ANALYSIS CALLBACK
# ==================================================

def analyze_company_dashboard(company):

    overview = ""
    analysis = ""

    try:
        overview = get_company_summary(company)

    except Exception as e:
        overview = (
            "### Company Summary Error\n"
            f"```text\n{str(e)}\n```"
        )

    try:
        analysis = generate_company_analysis(company)

    except Exception as e:
        analysis = (
            "### AI Analysis Error\n"
            f"```text\n{str(e)}\n```"
        )

    return str(overview), str(analysis)


# ==================================================
# 3. RAG FINANCIAL RESEARCH
# ==================================================

def run_financial_rag(question):

    if not question or not question.strip():
        return "Please enter a financial question.", pd.DataFrame()

    prompt = f"""
    You are an evidence-based financial research assistant.

    Use only the retrieved indexed documents.

    RULES:
    - Never invent financial figures.
    - Do not confuse companies or their financial data.
    - Use "one-year price return", not total return.
    - P/E alone does not establish undervaluation.
    - Do not interpret debt-to-equity without confirmed units.
    - RSI is an indicator, not a guaranteed prediction.
    - Separate facts from interpretations.
    - Identify missing information.
    - Never provide guaranteed predictions.

    Structure:
    1. Financial Strengths
    2. Financial Weaknesses and Risks
    3. Information Not Available
    4. Overall Assessment

    Question:
    {question}
    """

    response = loaded_query_engine.query(prompt)

    answer = re.sub(
        r"one-year\s+total\s+return",
        "one-year price return",
        response.response,
        flags=re.IGNORECASE
    )

    sources = []

    for source in response.source_nodes:
        sources.append({
            "Company": source.node.metadata.get("company"),
            "Document Type": source.node.metadata.get("document_type"),
            "Relevance Score": round(source.score, 3)
        })

    return answer, pd.DataFrame(sources)


# ==================================================
# 4. CREATE DASHBOARD
# ==================================================

with gr.Blocks(
    title="AI Financial Stock Analyzer",
    theme=gr.themes.Soft()
) as dashboard:

    gr.Markdown(
        """
        # AI Financial Stock Analyzer

        ### Powered by Groq, LlamaIndex, RAG and Yahoo Finance

        Analyze companies, compare competitors, explore financial
        news and generate AI-powered financial research.
        """
    )

    # ==============================================
    # TAB 1: COMPANY ANALYSIS
    # ==============================================

    with gr.Tab("Company Analysis"):

        gr.Markdown("## Individual Company Analysis")

        company_input = gr.Dropdown(
            choices=["NVIDIA", "AMD", "Intel"],
            value="NVIDIA",
            label="Select Company"
        )

        analyze_button = gr.Button(
            "Analyze Company",
            variant="primary"
        )

        company_overview = gr.Markdown()

        company_ai_analysis = gr.Markdown()

        # One combined callback instead of two click events
        analyze_button.click(
            fn=analyze_company_dashboard,
            inputs=company_input,
            outputs=[
                company_overview,
                company_ai_analysis
            ]
        )

    # ==============================================
    # TAB 2: COMPANY COMPARISON
    # ==============================================

    with gr.Tab("Company Comparison"):

        gr.Markdown("## Competitor Financial Comparison")

        gr.Dataframe(
            value=comparison_df,
            label="Company Financial Comparison",
            interactive=False
        )

        if "Revenue" in comparison_df.columns:
            gr.BarPlot(
                value=comparison_df,
                x="Company",
                y="Revenue",
                title="Revenue by Company"
            )

        if "Revenue Growth (%)" in comparison_df.columns:
            gr.BarPlot(
                value=comparison_df,
                x="Company",
                y="Revenue Growth (%)",
                title="Revenue Growth (%)"
            )

        if "Gross Margin (%)" in comparison_df.columns:
            gr.BarPlot(
                value=comparison_df,
                x="Company",
                y="Gross Margin (%)",
                title="Gross Margin (%)"
            )

        if "One-Year Return (%)" in comparison_df.columns:
            gr.BarPlot(
                value=comparison_df,
                x="Company",
                y="One-Year Return (%)",
                title="One-Year Price Return (%)"
            )

    # ==============================================
    # TAB 3: FINANCIAL NEWS ANALYSIS
    # ==============================================

    with gr.Tab("Financial News Analysis"):

        gr.Markdown("## Financial News and Sentiment")

        news_company = gr.Dropdown(
            choices=["NVIDIA", "AMD", "Intel"],
            value="NVIDIA",
            label="Select Company"
        )

        news_button = gr.Button(
            "Load Financial News",
            variant="primary"
        )

        company_news_table = gr.Dataframe(
            label="Company Financial News",
            interactive=False
        )

        news_analysis_output = gr.Markdown()

        news_button.click(
            fn=display_company_news,
            inputs=news_company,
            outputs=company_news_table
        )

        news_button.click(
            fn=generate_news_analysis,
            inputs=news_company,
            outputs=news_analysis_output
        )

        gr.Markdown("### Overall News Sentiment")

        gr.Dataframe(
            value=sentiment_summary.reset_index(),
            label="Sentiment Summary",
            interactive=False
        )

        gr.Plot(
            value=create_sentiment_chart(),
            label="News Sentiment Distribution"
        )

    # ==============================================
    # TAB 4: AI FINANCIAL RESEARCH (RAG)
    # ==============================================

    with gr.Tab("AI Financial Research (RAG)"):

        gr.Markdown(
            """
            ## Ask Your Financial AI Assistant

            Ask questions about the financial documents
            indexed using LlamaIndex and Groq.
            """
        )

        rag_question = gr.Textbox(
            label="Enter Your Financial Question",
            placeholder=(
                "Example: What are NVIDIA's financial "
                "strengths and weaknesses?"
            ),
            lines=3
        )

        rag_button = gr.Button(
            "Generate AI Research",
            variant="primary"
        )

        rag_answer = gr.Markdown()

        rag_sources = gr.Dataframe(
            label="Retrieved Source Documents",
            interactive=False
        )

        rag_button.click(
            fn=run_financial_rag,
            inputs=rag_question,
            outputs=[
                rag_answer,
                rag_sources
            ]
        )

    # ==============================================
    # FOOTER
    # ==============================================

    gr.Markdown(
        """
        ---

        **AI Financial Stock Analyzer**

        Powered by Yahoo Finance, Groq, LlamaIndex,
        Hugging Face Embeddings and Gradio.

        *For financial research and educational purposes only.*
        """
    )


# ==================================================
# 5. LAUNCH
# ==================================================

print("AI Financial Stock Analyzer is ready!")

dashboard.launch(share=True)
````

````python

import gradio as gr
import pandas as pd
import re
import traceback

# ==================================================
# 1. PREPARE COMPANY COMPARISON DATA
# ==================================================

comparison_df = pd.DataFrame.from_dict(
    competitor_data,
    orient="index"
).reset_index()

comparison_df = comparison_df.rename(
    columns={"index": "Company"}
)

# ==================================================
# 2. COMPANY ANALYSIS CALLBACK
# ==================================================

def analyze_company_dashboard(company):

    try:
        overview = get_company_summary(company)
    except Exception as e:
        overview = f"### Summary Error\n{str(e)}"

    try:
        analysis = generate_company_analysis(company)
    except Exception as e:
        analysis = f"### AI Analysis Error\n{str(e)}"

    return str(overview), str(analysis)


# ==================================================
# 3. FINANCIAL NEWS CALLBACK
# ==================================================

def analyze_news_dashboard(company):

    try:
        news_report = display_company_news(company)
    except Exception as e:
        news_report = f"### News Error\n{str(e)}"

    try:
        news_analysis = generate_news_analysis(company)
    except Exception as e:
        news_analysis = f"### News Analysis Error\n{str(e)}"

    return str(news_report), str(news_analysis)


# ==================================================
# 4. RAG FINANCIAL RESEARCH CALLBACK
# ==================================================

def run_financial_rag(question):

    if not question or not question.strip():
        return "Please enter a financial question.", pd.DataFrame()

    prompt = f"""
    You are an evidence-based financial research assistant.

    Use only the retrieved indexed documents.

    RULES:
    1. Never invent financial figures.
    2. Never confuse data belonging to different companies.
    3. Use "one-year price return", not total return.
    4. P/E ratios alone do not establish undervaluation.
    5. Do not interpret debt-to-equity without confirmed units.
    6. RSI is an indicator, not a guaranteed prediction.
    7. Separate reported facts from interpretations.
    8. Clearly identify missing information.
    9. Never provide guaranteed stock predictions.

    Structure:
    - Financial Strengths
    - Financial Weaknesses and Risks
    - Information Not Available
    - Overall Assessment

    Question:
    {question}
    """

    try:
        response = loaded_query_engine.query(prompt)

        answer = re.sub(
            r"one-year\s+total\s+return",
            "one-year price return",
            response.response,
            flags=re.IGNORECASE
        )

        source_records = []

        for source in response.source_nodes:
            source_records.append({
                "Company": source.node.metadata.get("company"),
                "Document Type": source.node.metadata.get("document_type"),
                "Relevance Score": round(source.score, 3)
            })

        return answer, pd.DataFrame(source_records)

    except Exception as e:
        return f"### RAG Error\n{str(e)}", pd.DataFrame()


# ==================================================
# 5. CREATE GRADIO DASHBOARD
# ==================================================

with gr.Blocks(
    title="AI Financial Stock Analyzer",
    theme=gr.themes.Soft()
) as dashboard:

    gr.Markdown(
        """
        # AI Financial Stock Analyzer

        ### Powered by Groq, LlamaIndex, RAG and Yahoo Finance

        Analyze company fundamentals, compare competitors,
        explore financial news and generate AI-powered research.
        """
    )

    # ==============================================
    # TAB 1: COMPANY ANALYSIS
    # ==============================================

    with gr.Tab("Company Analysis"):

        gr.Markdown("## Individual Company Analysis")

        company_input = gr.Dropdown(
            choices=["NVIDIA", "AMD", "Intel"],
            value="NVIDIA",
            label="Select Company"
        )

        analyze_button = gr.Button(
            "Analyze Company",
            variant="primary"
        )

        # Both functions return Markdown strings
        company_overview = gr.Markdown(
            label="Financial Overview"
        )

        company_ai_analysis = gr.Markdown(
            label="AI Company Analysis"
        )

        analyze_button.click(
            fn=analyze_company_dashboard,
            inputs=company_input,
            outputs=[
                company_overview,
                company_ai_analysis
            ]
        )

    # ==============================================
    # TAB 2: COMPANY COMPARISON
    # ==============================================

    with gr.Tab("Company Comparison"):

        gr.Markdown("## Competitor Financial Comparison")

        gr.Dataframe(
            value=comparison_df,
            label="Company Financial Comparison",
            interactive=False
        )

        if "Revenue" in comparison_df.columns:
            gr.BarPlot(
                value=comparison_df,
                x="Company",
                y="Revenue",
                title="Revenue by Company"
            )

        if "Revenue Growth (%)" in comparison_df.columns:
            gr.BarPlot(
                value=comparison_df,
                x="Company",
                y="Revenue Growth (%)",
                title="Revenue Growth (%)"
            )

        if "Gross Margin (%)" in comparison_df.columns:
            gr.BarPlot(
                value=comparison_df,
                x="Company",
                y="Gross Margin (%)",
                title="Gross Margin (%)"
            )

        if "One-Year Return (%)" in comparison_df.columns:
            gr.BarPlot(
                value=comparison_df,
                x="Company",
                y="One-Year Return (%)",
                title="One-Year Price Return (%)"
            )

    # ==============================================
    # TAB 3: FINANCIAL NEWS ANALYSIS
    # ==============================================

    with gr.Tab("Financial News Analysis"):

        gr.Markdown("## Financial News and Sentiment")

        news_company = gr.Dropdown(
            choices=["NVIDIA", "AMD", "Intel"],
            value="NVIDIA",
            label="Select Company"
        )

        news_button = gr.Button(
            "Load Financial News",
            variant="primary"
        )

        # FIX: display_company_news returns a string
        company_news_output = gr.Markdown(
            label="Company Financial News"
        )

        news_analysis_output = gr.Markdown(
            label="AI News Analysis"
        )

        # One callback handles both text outputs
        news_button.click(
            fn=analyze_news_dashboard,
            inputs=news_company,
            outputs=[
                company_news_output,
                news_analysis_output
            ]
        )

        gr.Markdown("### Overall News Sentiment")

        gr.Dataframe(
            value=sentiment_summary.reset_index(),
            label="Sentiment Summary",
            interactive=False
        )

        gr.Plot(
            value=create_sentiment_chart(),
            label="News Sentiment Distribution"
        )

    # ==============================================
    # TAB 4: AI FINANCIAL RESEARCH (RAG)
    # ==============================================

    with gr.Tab("AI Financial Research (RAG)"):

        gr.Markdown(
            """
            ## Ask Your Financial AI Assistant

            Powered by LlamaIndex, Groq and your
            indexed financial documents.
            """
        )

        rag_question = gr.Textbox(
            label="Enter Your Financial Question",
            placeholder=(
                "Example: What are NVIDIA's financial "
                "strengths and weaknesses?"
            ),
            lines=3
        )

        rag_button = gr.Button(
            "Generate AI Research",
            variant="primary"
        )

        rag_answer = gr.Markdown(
            label="AI Financial Research Report"
        )

        rag_sources = gr.Dataframe(
            label="Retrieved Source Documents",
            interactive=False
        )

        rag_button.click(
            fn=run_financial_rag,
            inputs=rag_question,
            outputs=[
                rag_answer,
                rag_sources
            ]
        )

    # ==============================================
    # FOOTER
    # ==============================================

    gr.Markdown(
        """
        ---

        **AI Financial Stock Analyzer**

        Powered by Yahoo Finance, Groq, LlamaIndex,
        Hugging Face Embeddings and Gradio.

        *For financial research and educational purposes only.*
        """
    )


# ==================================================
# 6. LAUNCH DASHBOARD
# ==================================================

print("AI Financial Stock Analyzer is ready!")

dashboard.launch(share=True)
````

````python

from google.colab import files
import os
import shutil

PROJECT_DIR = "/content/medical-chatbot"

# Create project folders
for folder in ["Data", "src", "templates", "static"]:
    os.makedirs(os.path.join(PROJECT_DIR, folder), exist_ok=True)

# Upload PDF directly from your computer
uploaded = files.upload()

# Get actual uploaded filename
pdf_files = [
    name for name in uploaded
    if name.lower().endswith(".pdf")
]

if not pdf_files:
    raise ValueError("No PDF uploaded. Please select Medical_book.pdf.")

pdf_name = pdf_files[0]

# Save PDF in project Data folder
destination = os.path.join(
    PROJECT_DIR, "Data", "Medical_book.pdf"
)

with open(destination, "wb") as f:
    f.write(uploaded[pdf_name])

print("PDF uploaded successfully!")
print("Saved at:", destination)
print("File size:", round(os.path.getsize(destination) / (1024 * 1024), 2), "MB")
````

````python

import os
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

PROJECT_DIR = "/content/medical-chatbot"
PDF_PATH = os.path.join(PROJECT_DIR, "Data", "Medical_book.pdf")

# Load medical PDF
loader = PyPDFLoader(PDF_PATH)
documents = loader.load()

print("PDF loaded successfully!")
print("Total pages:", len(documents))

# Split extracted text into chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=20
)

text_chunks = text_splitter.split_documents(documents)

print("Total text chunks:", len(text_chunks))
print("\nSample extracted text:\n")
print(text_chunks[0].page_content[:500])
````

````python

from langchain_community.embeddings import HuggingFaceEmbeddings

# Load the embedding model
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)

# Generate an embedding for one sample chunk
sample_text = text_chunks[0].page_content
sample_vector = embeddings.embed_query(sample_text)

print("Embedding model loaded successfully!")
print("Vector dimensions:", len(sample_vector))
print("Sample vector:", sample_vector[:10])
````

````python

from langchain_huggingface import HuggingFaceEmbeddings

# Initialize Hugging Face embeddings
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)

# Test embedding generation
sample_text = text_chunks[0].page_content
sample_vector = embeddings.embed_query(sample_text)

print("Embedding model initialized successfully!")
print("Model: all-MiniLM-L6-v2")
print("Vector dimensions:", len(sample_vector))
print("Total document chunks:", len(text_chunks))
````

````python

from pinecone import Pinecone, ServerlessSpec

# Connect to Pinecone
pc = Pinecone(api_key=PINECONE_API_KEY)

INDEX_NAME = "medicalbot"

# Check whether the index already exists
existing_indexes = pc.list_indexes().names()

if INDEX_NAME not in existing_indexes:
    pc.create_index(
        name=INDEX_NAME,
        dimension=384,
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
    print("Pinecone index created successfully!")
else:
    print("Pinecone index already exists.")

# Verify index configuration
index_description = pc.describe_index(INDEX_NAME)

print("\nIndex name:", index_description.name)
print("Index dimension:", index_description.dimension)
print("Index status:", index_description.status)
````

````python

import os

# Register the existing API key as an environment variable
os.environ["PINECONE_API_KEY"] = PINECONE_API_KEY

# Verify without displaying the secret
print(
    "Pinecone environment configured:",
    bool(os.environ.get("PINECONE_API_KEY"))
)
````

````python

import os
from langchain_pinecone import PineconeVectorStore

# Ensure Pinecone API key is available
assert os.environ.get("PINECONE_API_KEY"), "Pinecone key missing!"

# Connect to the existing index
docsearch = PineconeVectorStore.from_existing_index(
    index_name=INDEX_NAME,
    embedding=embeddings
)

print("Connected to Pinecone successfully!")
````

````python

# Upload medical chunks to Pinecone in batches

BATCH_SIZE = 100

for start in range(0, len(text_chunks), BATCH_SIZE):
    batch = text_chunks[start:start + BATCH_SIZE]

    # Stable IDs prevent duplicate records on reruns
    batch_ids = [
        f"medical_chunk_{i}"
        for i in range(start, start + len(batch))
    ]

    docsearch.add_documents(
        documents=batch,
        ids=batch_ids
    )

    print(
        f"Uploaded {min(start + len(batch), len(text_chunks))}"
        f" / {len(text_chunks)} chunks"
    )

# Verify Pinecone storage
index = pc.Index(INDEX_NAME)
stats = index.describe_index_stats()

print("\nUpload process finished!")
print("Vectors in Pinecone:", stats.total_vector_count)
````

````python
from pinecone import Pinecone
from langchain_pinecone import PineconeVectorStore

pc = Pinecone(api_key=PINECONE_API_KEY)

INDEX_NAME = "medicalbot"

docsearch = PineconeVectorStore.from_existing_index(
    index_name=INDEX_NAME,
    embedding=embeddings
)

retriever = docsearch.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

print("Pinecone retriever connected!")
print("Total vectors:", pc.Index(INDEX_NAME).describe_index_stats().total_vector_count)
````

````python
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate

# Medical chatbot instructions
system_prompt = """
You are a medical information assistant.

Answer the user's question using ONLY the retrieved medical context.

Instructions:
1. Explain medical information in simple, understandable language.
2. Do not invent facts or information.
3. If the retrieved context does not contain the answer, say:
   "I don't have enough information in my medical reference to answer that."
4. Organize answers using headings or bullet points when appropriate.
5. Never claim to diagnose a patient or prescribe treatment.
6. Encourage consultation with a qualified healthcare professional.
7. The medical reference may be outdated. Clearly communicate this limitation.

Retrieved medical context:
{context}
"""

# Create chat prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}")
])

# Create document QA chain
question_answer_chain = create_stuff_documents_chain(
    llm,
    prompt
)

# Create complete RAG chain
rag_chain = create_retrieval_chain(
    retriever,
    question_answer_chain
)

print("Medical RAG pipeline created successfully!")
````

````python
# Test question
question = "What is the function of the human heart?"

# Get answer from RAG pipeline
response = rag_chain.invoke({
    "input": question
})

print("QUESTION:")
print(question)

print("\nMEDICAL CHATBOT ANSWER:")
print(response["answer"])

print("\nRETRIEVED SOURCE DOCUMENTS:")
for i, doc in enumerate(response["context"], 1):
    print(f"\nSource {i}:")
    print("Page:", doc.metadata.get("page", "Unknown"))
    print("Content:", doc.page_content[:300])
````

````python
def hybrid_retrieve(question, k=5):
    # 1. Semantic retrieval
    semantic_docs = docsearch.similarity_search(
        question,
        k=k
    )

    # 2. Keyword retrieval
    question_words = [
        word.lower().strip(".,?!")
        for word in question.split()
        if len(word) > 3
    ]

    keyword_docs = []

    for chunk in text_chunks:
        content = chunk.page_content.lower()

        if any(word in content for word in question_words):
            keyword_docs.append(chunk)

    # 3. Merge and remove duplicate chunks
    combined_docs = []
    seen = set()

    for doc in semantic_docs + keyword_docs:
        key = (
            doc.metadata.get("page"),
            doc.page_content
        )

        if key not in seen:
            seen.add(key)
            combined_docs.append(doc)

        if len(combined_docs) >= k:
            break

    return combined_docs


# Test hybrid retrieval
question = "What is cardiac output?"

hybrid_results = hybrid_retrieve(question, k=5)

print("HYBRID RETRIEVAL RESULTS\n")

for i, doc in enumerate(hybrid_results, 1):
    print(f"Source {i}")
    print("Page:", doc.metadata.get("page"))
    print("Content:", doc.page_content[:500])
    print("-" * 60)
````

````python
def improved_hybrid_retrieve(question, k=5):

    # Extract meaningful query words
    stopwords = {
        "what", "is", "the", "of", "a", "an",
        "does", "how", "which", "are", "and"
    }

    words = [
        word.strip(".,?!").lower()
        for word in question.split()
        if word.strip(".,?!").lower() not in stopwords
    ]

    # 1. Exact phrase and keyword search
    keyword_results = []

    for chunk in text_chunks:
        content = chunk.page_content.lower()

        exact_phrase = " ".join(words)

        score = 0

        if exact_phrase in content:
            score += 10

        for word in words:
            if word in content:
                score += 1

        if score > 0:
            keyword_results.append((score, chunk))

    # Rank keyword matches
    keyword_results.sort(
        key=lambda item: item[0],
        reverse=True
    )

    # 2. Semantic search
    semantic_results = docsearch.similarity_search(
        question,
        k=k
    )

    # 3. Merge keyword results first, then semantic results
    combined = []
    seen = set()

    for doc in (
        [item[1] for item in keyword_results]
        + semantic_results
    ):
        key = (
            doc.metadata.get("page"),
            doc.page_content
        )

        if key not in seen:
            seen.add(key)
            combined.append(doc)

        if len(combined) >= k:
            break

    return combined


# Test improved retrieval
question = "What is cardiac output?"

results = improved_hybrid_retrieve(question, k=5)

print("IMPROVED HYBRID RETRIEVAL\n")

for i, doc in enumerate(results, 1):
    print(f"Source {i}")
    print("Page:", doc.metadata.get("page"))
    print("Content:", doc.page_content[:500])
    print("-" * 60)
````

````python
import re

def improved_hybrid_retrieve(question, k=5):

    stopwords = {
        "what", "is", "the", "of", "a", "an",
        "does", "how", "which", "are", "and",
        "with", "for", "in", "to", "on", "by"
    }

    keywords = [
        word.lower()
        for word in re.findall(r"\b[a-zA-Z]+\b", question)
        if word.lower() not in stopwords
    ]

    exact_phrase = " ".join(keywords)

    # Collect keyword matches from the original PDF chunks
    candidates = []

    for doc in text_chunks:
        content = doc.page_content.lower()

        matched = [word for word in keywords if word in content]
        coverage = len(matched) / max(len(keywords), 1)
        phrase_match = exact_phrase in content

        # Keep only documents with strong keyword relevance
        if coverage >= 0.75:
            score = (coverage * 10) + (5 if phrase_match else 0)
            candidates.append((score, doc))

    # Sort strongest matches first
    candidates.sort(key=lambda item: item[0], reverse=True)

    # Add semantic matches, but apply the same relevance filter
    semantic_docs = docsearch.similarity_search(question, k=10)

    seen = set()
    results = []

    for score, doc in candidates:
        key = (doc.metadata.get("page"), doc.page_content)

        if key not in seen:
            seen.add(key)
            results.append(doc)

    for doc in semantic_docs:
        content = doc.page_content.lower()
        matched = [word for word in keywords if word in content]
        coverage = len(matched) / max(len(keywords), 1)

        if coverage >= 0.75:
            key = (doc.metadata.get("page"), doc.page_content)

            if key not in seen:
                seen.add(key)
                results.append(doc)

    return results[:k]


# Test the improved retrieval
question = "What is cardiac output?"
retrieved_docs = improved_hybrid_retrieve(question, k=5)

print("Question:", question)
print("Number of relevant documents:", len(retrieved_docs))

for i, doc in enumerate(retrieved_docs, 1):
    print(f"\nSource {i} | Page {doc.metadata.get('page')}")
    print(doc.page_content[:300])
````

````python
test_questions = {
    "Hypertension": "What is hypertension?",
    "Diabetes": "What are the symptoms of diabetes?",
    "Anemia": "What is anemia?",
    "Asthma": "What causes asthma?",
    "Cardiac Output": "What is cardiac output?"
}

import pandas as pd
import re

evaluation_results = []

for topic, question in test_questions.items():

    docs = improved_hybrid_retrieve(question, k=5)

    keywords = [
        word.lower()
        for word in re.findall(r"\b[a-zA-Z]+\b", question)
        if word.lower() not in {
            "what", "is", "the", "of", "a", "an",
            "does", "how", "which", "are", "and",
            "with", "for", "in", "to"
        }
    ]

    relevant_count = 0

    for doc in docs:
        content = doc.page_content.lower()

        if any(word in content for word in keywords):
            relevant_count += 1

    evaluation_results.append({
        "Topic": topic,
        "Retrieved Documents": len(docs),
        "Keyword Relevant Documents": relevant_count,
        "Source Pages": [
            doc.metadata.get("page") for doc in docs
        ]
    })

evaluation_df = pd.DataFrame(evaluation_results)

display(evaluation_df)
````

````python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2",
    device="cpu"
)

print("Cross-encoder reranking model loaded successfully!")
````

````python
def rerank_medical_documents(question, documents, top_k=3):

    pairs = [
        [question, doc.page_content]
        for doc in documents
    ]

    scores = reranker.predict(pairs)

    ranked_docs = sorted(
        zip(scores, documents),
        key=lambda item: item[0],
        reverse=True
    )

    results = []

    for score, doc in ranked_docs[:top_k]:
        results.append({
            "score": float(score),
            "document": doc
        })

    return results


print("Medical document reranking function created successfully!")
````

````python
for topic, question in test_questions.items():

    print("\n" + "=" * 75)
    print("TOPIC:", topic)
    print("QUESTION:", question)

    # Retrieve candidate documents
    candidates = improved_hybrid_retrieve(question, k=10)

    # Rerank candidates
    ranked_results = rerank_medical_documents(
        question,
        candidates,
        top_k=3
    )

    print("\nTOP 3 RERANKED DOCUMENTS:")

    for i, item in enumerate(ranked_results, 1):
        doc = item["document"]

        print(f"\nRank: {i}")
        print(f"Relevance score: {item['score']:.4f}")
        print(f"Page: {doc.metadata.get('page')}")
        print("Content:", doc.page_content[:400])

    print("=" * 75)
````

````python
def reranked_medical_retrieve(question, candidate_k=10, top_k=3):

    # Stage 1: Hybrid candidate retrieval
    candidates = improved_hybrid_retrieve(
        question,
        k=candidate_k
    )

    if not candidates:
        return []

    # Stage 2: Cross-encoder reranking
    ranked_results = rerank_medical_documents(
        question,
        candidates,
        top_k=top_k
    )

    # Return documents in ranked order
    return [
        item["document"]
        for item in ranked_results
    ]


print("Hybrid retrieval + cross-encoder reranking pipeline created!")
````

````python
def improved_medical_chat(question):

    # Retrieve and rerank relevant medical passages
    retrieved_docs = reranked_medical_retrieve(
        question,
        candidate_k=10,
        top_k=3
    )

    if not retrieved_docs:
        return {
            "answer": "I could not find relevant information in the supplied medical reference.",
            "sources": []
        }

    # Prepare context with page references
    context = "\n\n".join(
        f"Source page: {doc.metadata.get('page')}\n{doc.page_content}"
        for doc in retrieved_docs
    )

    messages = [
        (
            "system",
            """You are a medical information assistant.

Answer using ONLY the supplied medical reference.

Rules:
1. Do not invent facts or information.
2. Answer the user's actual question, not merely related topics.
3. Explain in simple language.
4. If the context does not contain the requested information, clearly say so.
5. Do not diagnose diseases or prescribe treatment.
6. Preserve the meaning and terminology of the supplied reference.
7. Include source page numbers in your answer.
8. Mention that the medical reference may be outdated.
9. Clearly distinguish information in the reference from information it does not provide.

Medical context:
{context}
""".format(context=context)
        ),
        ("human", question)
    ]

    response = llm.invoke(messages)

    return {
        "answer": response.content,
        "sources": retrieved_docs
    }


print("Improved Groq medical chatbot created successfully!")
````

````python
%%writefile /content/medical-chatbot/src/helper.py

import os
import re

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_pinecone import PineconeVectorStore

from pinecone import Pinecone
from sentence_transformers import CrossEncoder


# Configuration
INDEX_NAME = "medicalbot"
PDF_PATH = os.path.join(
    os.path.dirname(os.path.dirname(__file__)),
    "Data",
    "Medical_book.pdf"
)

EMBEDDING_MODEL = "sentence-transformers/all-MiniLM-L6-v2"
RERANKER_MODEL = "cross-encoder/ms-marco-MiniLM-L-6-v2"


def initialize_components():

    pinecone_key = os.getenv("PINECONE_API_KEY")

    if not pinecone_key:
        raise ValueError("PINECONE_API_KEY is missing.")

    if not os.path.exists(PDF_PATH):
        raise FileNotFoundError(f"Medical PDF not found: {PDF_PATH}")

    print("Loading Hugging Face embeddings...")

    embeddings = HuggingFaceEmbeddings(
        model_name=EMBEDDING_MODEL,
        model_kwargs={"device": "cpu"},
        encode_kwargs={"normalize_embeddings": True}
    )

    print("Connecting to existing Pinecone index...")

    pc = Pinecone(api_key=pinecone_key)

    index = pc.Index(INDEX_NAME)

    stats = index.describe_index_stats()

    if stats.total_vector_count == 0:
        raise ValueError("The Pinecone index contains no vectors.")

    docsearch = PineconeVectorStore.from_existing_index(
        index_name=INDEX_NAME,
        embedding=embeddings
    )

    print("Loading medical PDF for keyword retrieval...")

    loader = PyPDFLoader(PDF_PATH)
    pages = loader.load()

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=20
    )

    text_chunks = splitter.split_documents(pages)

    print("Loading cross-encoder reranker...")

    reranker = CrossEncoder(
        RERANKER_MODEL,
        device="cpu"
    )

    print("All components initialized successfully!")

    return docsearch, text_chunks, reranker


def hybrid_retrieve(question, docsearch, text_chunks, reranker,
                    candidate_k=10, top_k=3):

    stopwords = {
        "what", "is", "the", "of", "a", "an",
        "does", "how", "which", "are", "and",
        "with", "for", "in", "to", "on", "by"
    }

    keywords = [
        word.lower()
        for word in re.findall(r"\b[a-zA-Z]+\b", question)
        if word.lower() not in stopwords
    ]

    candidates = []
    seen = set()

    # Keyword retrieval
    for doc in text_chunks:

        content = doc.page_content.lower()

        matched = [
            word for word in keywords
            if word in content
        ]

        coverage = len(matched) / max(len(keywords), 1)

        if coverage >= 0.75:

            key = (doc.metadata.get("page"), doc.page_content)

            if key not in seen:
                seen.add(key)
                candidates.append(doc)

    # Semantic retrieval from Pinecone
    semantic_docs = docsearch.similarity_search(
        question,
        k=candidate_k
    )

    for doc in semantic_docs:

        key = (doc.metadata.get("page"), doc.page_content)

        if key not in seen:
            seen.add(key)
            candidates.append(doc)

    if not candidates:
        return []

    # Cross-encoder reranking
    pairs = [
        [question, doc.page_content]
        for doc in candidates
    ]

    scores = reranker.predict(pairs)

    ranked_docs = sorted(
        zip(scores, candidates),
        key=lambda item: item[0],
        reverse=True
    )

    # Remove duplicate passages and pages
    results = []
    seen_content = set()

    for score, doc in ranked_docs:

        normalized_content = re.sub(
            r"\s+", " ", doc.page_content
        ).strip().lower()

        if normalized_content in seen_content:
            continue

        seen_content.add(normalized_content)

        results.append({
            "score": float(score),
            "document": doc
        })

        if len(results) >= top_k:
            break

    return results
````

````python
%%writefile /content/medical-chatbot/src/prompt.py

SYSTEM_PROMPT = """
You are a medical information assistant.

Your task is to answer user questions using the supplied medical reference.

STRICT RULES:

1. Use ONLY the information supported by the supplied context.
2. Do not invent medical facts, symptoms, causes, diagnoses, or treatments.
3. Answer the exact question asked by the user.
4. Explain medical concepts in simple and understandable language.
5. If the context does not contain enough information, clearly say so.
6. Do not prescribe medicines, dosages, or treatment plans.
7. Do not claim that the information is current or clinically validated.
8. Preserve the terminology and meaning of the supplied reference.
9. Mention source page numbers when available.
10. Clearly distinguish direct information from the reference and information it does not provide.
11. Remind users that this reference may be outdated.
12. Encourage consultation with a qualified healthcare professional for personal medical concerns.

IMPORTANT:
If the retrieved passages discuss a related condition but do not answer the actual question, do not present them as the answer.

MEDICAL CONTEXT:

{context}
"""

HUMAN_PROMPT = """
User question:

{question}

Provide a clear, concise, source-supported answer.
"""
````

````python

%%writefile /content/medical-chatbot/app.py

import os
import re
import logging

from flask import Flask, render_template, request, jsonify
from dotenv import load_dotenv
from langchain_groq import ChatGroq

from src.helper import initialize_components, hybrid_retrieve
from src.prompt import SYSTEM_PROMPT, HUMAN_PROMPT

load_dotenv()

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

app = Flask(__name__)

# Initialize components
logger.info("Initializing Medical AI Assistant...")

docsearch, text_chunks, reranker = initialize_components()

GROQ_API_KEY = os.getenv("GROQ_API_KEY")

if not GROQ_API_KEY:
    raise ValueError("GROQ_API_KEY is missing.")

llm = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0.2,
    max_tokens=700,
    api_key=GROQ_API_KEY
)

logger.info("Medical AI Assistant initialized successfully!")


# Clean escaped Markdown
def clean_markdown(text):

    text = re.sub(r'\\([*_`#])', r'\1', text)

    return text.strip()


# Generate medical response
def get_medical_response(question):

    question = question.strip()

    if not question:
        return {
            "answer": "Please enter a medical question.",
            "sources": []
        }

    retrieved_results = hybrid_retrieve(
        question=question,
        docsearch=docsearch,
        text_chunks=text_chunks,
        reranker=reranker,
        candidate_k=10,
        top_k=3
    )

    if not retrieved_results:
        return {
            "answer": (
                "I could not find sufficient relevant information "
                "in the supplied medical reference."
            ),
            "sources": []
        }

    context_parts = []
    sources = []

    for item in retrieved_results:

        doc = item["document"]
        page = doc.metadata.get("page")

        page_number = (
            int(float(page)) + 1
            if page is not None else "Unknown"
        )

        context_parts.append(
            f"Source page: {page_number}\n{doc.page_content}"
        )

        sources.append({
            "page": page_number,
            "relevance_score": round(float(item["score"]), 4)
        })

    context = "\n\n".join(context_parts)

    response = llm.invoke([
        (
            "system",
            SYSTEM_PROMPT.format(context=context)
        ),
        (
            "human",
            HUMAN_PROMPT.format(question=question)
        )
    ])

    answer = clean_markdown(response.content)

    return {
        "answer": answer,
        "sources": sources
    }


@app.route("/")
def index():
    return render_template("chat.html")


@app.route("/get", methods=["GET", "POST"])
def chat():

    question = (
        request.form.get("msg")
        or request.form.get("question")
        or request.args.get("msg")
        or request.args.get("question")
        or ""
    )

    if not question.strip():
        return "Please enter a medical question.", 400

    try:
        result = get_medical_response(question)
        return result["answer"]

    except Exception:
        logger.exception("Chat request failed")
        return "An error occurred while processing your question.", 500


@app.route("/api/chat", methods=["POST"])
def api_chat():

    data = request.get_json(silent=True) or {}
    question = data.get("question", "")

    if not isinstance(question, str) or not question.strip():
        return jsonify({"error": "Please enter a valid question."}), 400

    if len(question) > 2000:
        return jsonify({"error": "Question is too long."}), 400

    try:
        return jsonify(get_medical_response(question))

    except Exception:
        logger.exception("Medical API request failed")
        return jsonify({
            "error": "An error occurred. Please try again."
        }), 500


@app.route("/health")
def health():

    return jsonify({
        "status": "running",
        "application": "Medical AI Assistant"
    })


if __name__ == "__main__":

    app.run(
        host="0.0.0.0",
        port=5000,
        debug=False,
        use_reloader=False,
        threaded=True
    )
````

````python

%%writefile /content/medical-chatbot/templates/chat.html

<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Medical AI Assistant</title>

    <link rel="stylesheet"
          href="{{ url_for('static', filename='style.css') }}">

    <script src="https://cdn.jsdelivr.net/npm/marked/marked.min.js"></script>
    <script src="https://cdn.jsdelivr.net/npm/dompurify@3/dist/purify.min.js"></script>
</head>

<body>

<div class="app-container">

    <header class="chat-header">
        <div class="logo">✚</div>

        <div class="header-info">
            <h1>Medical AI Assistant</h1>
            <p>AI-powered medical information assistant</p>
        </div>

        <div class="header-actions">
            <span class="status">
                <span class="status-dot"></span>
                Online
            </span>

            <button id="clear-chat" class="clear-button" type="button">
                Clear chat
            </button>
        </div>
    </header>


    <main class="chat-area" id="chat-area">

        <div class="welcome-screen" id="welcome-screen">

            <div class="welcome-icon">✚</div>

            <h2>How can I help you today?</h2>

            <p>
                Ask educational questions about medical conditions,
                symptoms and healthcare topics.
            </p>

            <div class="suggestion-list">

                <button class="suggestion"
                        data-question="What is hypertension?">
                    What is hypertension?
                </button>

                <button class="suggestion"
                        data-question="What causes asthma?">
                    Causes of asthma
                </button>

                <button class="suggestion"
                        data-question="What is anemia?">
                    Explain anemia
                </button>

                <button class="suggestion"
                        data-question="What is diabetes?">
                    What is diabetes?
                </button>

            </div>
        </div>

    </main>


    <div class="loading" id="loading">
        <div class="loading-dots">
            <span></span>
            <span></span>
            <span></span>
        </div>
        <span>Searching medical reference...</span>
    </div>


    <form id="chat-form" class="chat-input-area">

        <input
            type="text"
            id="user-input"
            placeholder="Ask your medical question..."
            autocomplete="off"
            required
        >

        <button type="submit" id="send-button">
            Send
        </button>

    </form>


    <footer>
        <p>Powered by Groq, Pinecone and Hugging Face</p>
        <p class="medical-disclaimer">
            Educational information only. Not a substitute for professional medical advice.
        </p>
    </footer>

</div>


<script>

marked.setOptions({
    breaks: true,
    gfm: true
});

const chatForm = document.getElementById("chat-form");
const userInput = document.getElementById("user-input");
const chatArea = document.getElementById("chat-area");
const loading = document.getElementById("loading");
const sendButton = document.getElementById("send-button");
const clearButton = document.getElementById("clear-chat");
const welcomeScreen = document.getElementById("welcome-screen");


// Normalize escaped Markdown before rendering
function cleanMarkdown(text) {

    return text
        .replace(/\\([*_`#])/g, "$1")
        .replace(/\r\n/g, "\n");

}


// Add a message to the conversation
function addMessage(text, sender, sources = []) {

    const message = document.createElement("div");
    message.className = "message " + sender + "-message";

    const avatar = document.createElement("div");
    avatar.className = "avatar";
    avatar.textContent = sender === "user" ? "You" : "AI";

    const content = document.createElement("div");
    content.className = "message-content";

    if (sender === "assistant") {

        const cleanedText = cleanMarkdown(text);

        const renderedHTML = marked.parse(cleanedText);

        content.innerHTML = DOMPurify.sanitize(renderedHTML);

    } else {

        const paragraph = document.createElement("p");
        paragraph.textContent = text;
        content.appendChild(paragraph);

    }


    // Expandable source references
    if (sender === "assistant" && sources.length > 0) {

        const details = document.createElement("details");
        details.className = "source-details";

        const summary = document.createElement("summary");

        summary.textContent =
            "View medical sources (" + sources.length + ")";

        const sourceList = document.createElement("ul");
        sourceList.className = "source-list";

        sources.forEach(source => {

            const item = document.createElement("li");

            item.textContent =
                "Medical reference — Page " + source.page;

            sourceList.appendChild(item);

        });

        details.appendChild(summary);
        details.appendChild(sourceList);

        content.appendChild(details);

    }


    message.appendChild(avatar);
    message.appendChild(content);

    chatArea.appendChild(message);

    chatArea.scrollTop = chatArea.scrollHeight;

}


// Hide welcome screen
function hideWelcomeScreen() {
    if (welcomeScreen) {
        welcomeScreen.style.display = "none";
    }
}


// Send question to Flask
async function sendQuestion(question) {

    question = question.trim();

    if (!question || sendButton.disabled) return;

    hideWelcomeScreen();

    addMessage(question, "user");

    userInput.value = "";

    loading.style.display = "flex";
    sendButton.disabled = true;

    try {

        const response = await fetch("/api/chat", {
            method: "POST",
            headers: {
                "Content-Type": "application/json"
            },
            body: JSON.stringify({
                question: question
            })
        });

        const data = await response.json();

        if (!response.ok) {
            throw new Error(data.error || "Request failed");
        }

        addMessage(
            data.answer || "No answer was generated.",
            "assistant",
            data.sources || []
        );

    } catch (error) {

        addMessage(
            "Sorry, something went wrong. Please try again.",
            "assistant"
        );

        console.error("Chat error:", error);

    } finally {

        loading.style.display = "none";
        sendButton.disabled = false;
        userInput.focus();

    }

}


// Submit form
chatForm.addEventListener("submit", function(event) {

    event.preventDefault();
    sendQuestion(userInput.value);

});


// Suggested questions
document.querySelectorAll(".suggestion").forEach(button => {

    button.addEventListener("click", function() {

        sendQuestion(this.dataset.question);

    });

});


// Clear conversation
clearButton.addEventListener("click", function() {

    chatArea.innerHTML = "";

    chatArea.appendChild(welcomeScreen);

    welcomeScreen.style.display = "block";

    userInput.value = "";

});

</script>

</body>
</html>
````

````python
%%writefile /content/source-code-analysis/src/helper.py

import os
import shutil

from git import Repo
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings


def repo_ingestion(repo_url, repo_path="repo"):

    if os.path.exists(repo_path):
        shutil.rmtree(repo_path)

    print("Cloning GitHub repository...")

    Repo.clone_from(repo_url, repo_path)

    print("Repository cloned successfully!")


def load_repo(repo_path="repo"):

    documents = []

    ignored_dirs = {
        ".git",
        "venv",
        ".venv",
        "__pycache__",
        "node_modules",
        "site-packages",
        "dist",
        "build"
    }

    for root, dirs, files in os.walk(repo_path):

        dirs[:] = [
            directory for directory in dirs
            if directory not in ignored_dirs
        ]

        for file in files:

            if file.endswith(".py"):

                file_path = os.path.join(root, file)

                try:
                    with open(
                        file_path,
                        "r",
                        encoding="utf-8",
                        errors="ignore"
                    ) as f:
                        content = f.read()

                    if content.strip():
                        documents.append(
                            Document(
                                page_content=content,
                                metadata={
                                    "source": file_path,
                                    "filename": file
                                }
                            )
                        )

                except Exception as e:
                    print(f"Skipping {file_path}: {e}")

    print(f"Python files loaded: {len(documents)}")

    return documents


def text_splitter(documents):

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=2000,
        chunk_overlap=200
    )

    chunks = splitter.split_documents(documents)

    print(f"Code chunks created: {len(chunks)}")

    return chunks


def load_embedding():

    print("Loading Hugging Face embeddings...")

    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2",
        model_kwargs={"device": "cpu"},
        encode_kwargs={"normalize_embeddings": True}
    )

    print("Embedding model loaded successfully!")

    return embeddings
````

````python
from langchain_huggingface import HuggingFaceEmbeddings

def load_embedding():

    print("Loading Hugging Face embedding model...")

    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2",
        model_kwargs={"device": "cpu"},
        encode_kwargs={"normalize_embeddings": True}
    )

    print("Embedding model loaded successfully!")

    return embeddings


embeddings = load_embedding()

# Test embedding generation
test_vector = embeddings.embed_query(
    "What is a Python function?"
)

print("Embedding dimension:", len(test_vector))
print("Embedding test successful!")
````

````python
%%writefile /content/source-code-analysis/store_index.py

import os
import shutil

from src.helper import (
    repo_ingestion,
    load_repo,
    text_splitter,
    load_embedding
)

from langchain_chroma import Chroma


def create_vector_database(repo_url):

    repo_path = "repo"
    persist_directory = "db"

    # Clone repository
    repo_ingestion(repo_url, repo_path)

    # Load Python files
    documents = load_repo(repo_path)

    if not documents:
        raise ValueError(
            "No Python files found in the GitHub repository."
        )

    # Split source code into chunks
    chunks = text_splitter(documents)

    # Load Hugging Face embeddings
    embeddings = load_embedding()

    # Remove previous database
    if os.path.exists(persist_directory):
        shutil.rmtree(persist_directory)

    # Create ChromaDB
    vectordb = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings,
        persist_directory=persist_directory,
        collection_name="source_code"
    )

    print("\nVector database created successfully!")
    print("Total code chunks indexed:", len(chunks))

    return vectordb
````
